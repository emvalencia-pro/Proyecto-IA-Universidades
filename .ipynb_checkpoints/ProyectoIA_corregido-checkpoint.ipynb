## Bloque 1 — `build_ontologia.py`

Construye el grafo con RDFLib (`g.add(...)`) y serializa `ontologia.ttl`.

**[C2/C9]** Se agregan 6 propiedades que conectan `Beca`, `Convenio` y `Estudiante` con el resto de la ontología, y cada estudiante trae su perfil completo (área, nivel, promedio e idiomas).

```python
# -*- coding: utf-8 -*-
"""
build_ontologia.py
==================================================================
Script CONSTRUCTOR de la ontología del dominio "Recomendación de
universidades para intercambio académico" (Práctica 1 - IA).

Este script NO forma parte del sistema en operación: se ejecuta
UNA sola vez para generar el archivo canónico `ontologia.ttl`
(decisión de arquitectura §6.2 del Context.md). A partir de ahí,
el sistema real solo LEE ese .ttl; el razonador nunca reescribe
el archivo en disco.

Construye el grafo RDF/RDFS con `g.add(...)` (más rápido y menos
propenso a errores que escribir Turtle a mano) y lo serializa en
formato Turtle.

Cumple los mínimos del enunciado (§2.3):
  - >=10 clases  (rdfs:Class)
  - >=5  jerarquías de clase   (rdfs:subClassOf)
  - >=10 propiedades propias con rdfs:domain y rdfs:range
  - >=1  jerarquía de propiedad (rdfs:subPropertyOf)
  - rangos de tipo clase (URI) y literales tipados (XML-Schema)
  - vocabulario externo aceptado (FOAF, Dublin Core)
  - >=4 individuos por clase
  - rdfs:label en TODAS las clases y propiedades

[CORRECCIÓN C2/C9] Se agregan 6 propiedades que CONECTAN las clases
que antes estaban aisladas (Beca, Convenio, Estudiante):
  tieneBeca, amparadaPorConvenio        (Oferta -> Beca / Convenio)
  estudiaArea, cursaNivel,               (Estudiante -> Área / Nivel)
  tienePromedio, dominaIdioma            (Estudiante -> literales XSD)
Así el perfil del estudiante sale de la ontología y las becas y
convenios participan en reglas reales del sistema experto.
==================================================================
"""

from rdflib import Graph, Namespace, Literal
from rdflib.namespace import RDF, RDFS, XSD, FOAF, DCTERMS

# ------------------------------------------------------------------
# 1. NAMESPACES Y GRAFO
# ------------------------------------------------------------------
# Prefijo del proyecto (Contrato §4.1). Toda entidad propia vive aquí.
EX = Namespace("http://intercambios.unal.edu.co/")

g = Graph()
# Enlazamos prefijos para que el Turtle salga legible.
g.bind("ex", EX)
g.bind("foaf", FOAF)
g.bind("dct", DCTERMS)
g.bind("rdfs", RDFS)
g.bind("xsd", XSD)


# ------------------------------------------------------------------
# 2. FUNCIONES AUXILIARES
# ------------------------------------------------------------------
def add_class(name, label, sub_class_of=None):
    """Declara una rdfs:Class con su rdfs:label (en español).

    name         : nombre local de la clase (se resuelve contra EX).
    label        : significado humano de la clase (requisito rdfs:label).
    sub_class_of : URIRef de la superclase, o None. Genera rdfs:subClassOf.
    """
    cls = EX[name]
    g.add((cls, RDF.type, RDFS.Class))
    g.add((cls, RDFS.label, Literal(label, lang="es")))
    if sub_class_of is not None:
        g.add((cls, RDFS.subClassOf, sub_class_of))
    return cls


def add_property(name, label, domain, rango, sub_property_of=None):
    """Declara una rdf:Property con domain, range y rdfs:label.

    domain / rango : URIRef (clase o tipo XSD).
    sub_property_of: URIRef de la superpropiedad, o None -> rdfs:subPropertyOf.
    """
    prop = EX[name]
    g.add((prop, RDF.type, RDF.Property))
    g.add((prop, RDFS.label, Literal(label, lang="es")))
    g.add((prop, RDFS.domain, domain))
    g.add((prop, RDFS.range, rango))
    if sub_property_of is not None:
        g.add((prop, RDFS.subPropertyOf, sub_property_of))
    return prop


# ------------------------------------------------------------------
# 3. CLASES Y JERARQUÍAS (rdfs:Class + rdfs:subClassOf)
# ------------------------------------------------------------------
# --- Clases raíz del dominio ---
Oferta          = add_class("Oferta", "Oferta de intercambio académico")
Universidad     = add_class("Universidad", "Institución de educación superior")
Pais            = add_class("Pais", "País donde se ubica una universidad")
AreaConocimiento= add_class("AreaConocimiento", "Área de conocimiento o facultad")
NivelEstudios   = add_class("NivelEstudios", "Nivel de estudios (pregrado, posgrado, maestría, doctorado)")
Convenio        = add_class("Convenio", "Convenio de movilidad entre instituciones")
Beca            = add_class("Beca", "Beca o apoyo económico para la movilidad")

# --- Jerarquías de clase (>=5 rdfs:subClassOf) ---
# 1 y 2: tipos de oferta
OfertaInternacional = add_class("OfertaInternacional", "Oferta de intercambio internacional", sub_class_of=Oferta)
OfertaNacional      = add_class("OfertaNacional",      "Oferta de intercambio nacional",      sub_class_of=Oferta)
# 3: universidad de destino
UniversidadDestino  = add_class("UniversidadDestino",  "Universidad anfitriona / de destino", sub_class_of=Universidad)
# 4: estudiante como persona FOAF (usa vocabulario externo como superclase)
Estudiante          = add_class("Estudiante",          "Estudiante candidato a la movilidad", sub_class_of=FOAF.Person)
# 5: Ingeniería como subclase de área (habilita el caso 1b de razonamiento)
Ingenieria          = add_class("Ingenieria",          "Área de Ingeniería",                  sub_class_of=AreaConocimiento)


# ------------------------------------------------------------------
# 4. PROPIEDADES (>=10 propias, con domain/range y rdfs:label)
# ------------------------------------------------------------------
# --- Rangos de tipo CLASE (URIs) ---
ofrecidaPor   = add_property("ofrecidaPor",   "es ofrecida por la universidad", Oferta, Universidad)
ofreceEnPais  = add_property("ofreceEnPais",  "se ofrece en el país",           Oferta, Pais)
tieneArea     = add_property("tieneArea",     "corresponde al área de conocimiento", Oferta, AreaConocimiento)
dirigidaANivel= add_property("dirigidaANivel","está dirigida al nivel de estudios",  Oferta, NivelEstudios)
restringeArea = add_property("restringeArea", "está restringida al área",       Oferta, AreaConocimiento)
ubicadaEn     = add_property("ubicadaEn",     "está ubicada en el país",         Universidad, Pais)
# [C2] Nuevas relaciones que conectan Beca, Convenio y Estudiante con el resto
tieneBeca          = add_property("tieneBeca",          "ofrece la beca",                          Oferta, Beca)
amparadaPorConvenio= add_property("amparadaPorConvenio","está respaldada por el convenio",         Oferta, Convenio)
estudiaArea        = add_property("estudiaArea",        "el estudiante pertenece al área",         Estudiante, AreaConocimiento)
cursaNivel         = add_property("cursaNivel",         "el estudiante cursa el nivel de estudios",Estudiante, NivelEstudios)

# --- Rangos de tipo LITERAL (tipos XML-Schema) ---
tieneCupos    = add_property("tieneCupos",    "número de cupos disponibles",     Oferta, XSD.integer)
fechaCierre   = add_property("fechaCierre",   "fecha de cierre de la convocatoria", Oferta, XSD.date)
requiereIdioma= add_property("requiereIdioma","nivel de idioma requerido",       Oferta, XSD.string)
ofreceBeca    = add_property("ofreceBeca",    "indica si la oferta ofrece beca", Oferta, XSD.boolean)
# [C9] Perfil del estudiante en la ontología (antes se escribía a mano en Python)
tienePromedio = add_property("tienePromedio", "promedio académico (P.A.P.A.) del estudiante", Estudiante, XSD.decimal)
dominaIdioma  = add_property("dominaIdioma",  "idioma y nivel MCER que domina el estudiante", Estudiante, XSD.string)

# --- Jerarquía de propiedad (rdfs:subPropertyOf) ---
# Habilita el caso 3 de razonamiento: toda oferta ofrecida por una
# UniversidadDestino queda también ligada por la propiedad general ofrecidaPor.
ofrecidaPorUniversidadDestino = add_property(
    "ofrecidaPorUniversidadDestino",
    "es ofrecida por una universidad de destino",
    Oferta, UniversidadDestino,
    sub_property_of=ofrecidaPor,
)


# ------------------------------------------------------------------
# 5. INDIVIDUOS DE CLASES DE APOYO (>=4 por clase)
# ------------------------------------------------------------------
def lbl(uri, texto):
    """Atajo: agrega rdfs:label en español a un individuo."""
    g.add((uri, RDFS.label, Literal(texto, lang="es")))

# --- Países ---------------------------------------------------------
# 4 países se declaran EXPLÍCITAMENTE como ex:Pais (cumple el mínimo
# de individuos). Los restantes (noruega, colombia) se dejan SIN
# rdf:type a propósito: el razonador los tipará vía el range de
# ofreceEnPais/ubicadaEn -> demostración del "caso 2" (rango).
PAISES_EXPLICITOS = {
    "pais_francia":  "Francia",
    "pais_italia":   "Italia",
    "pais_mexico":   "México",
    "pais_alemania": "Alemania",
}
PAISES_INFERIDOS = {  # se tiparán por razonamiento (range); deben ser usados por alguna oferta
    "pais_noruega":  "Noruega",
    "pais_colombia": "Colombia",
}
for pid, nombre in PAISES_EXPLICITOS.items():
    g.add((EX[pid], RDF.type, Pais))
    lbl(EX[pid], nombre)                      # un país no es un agente FOAF: basta rdfs:label
for pid, nombre in PAISES_INFERIDOS.items():
    # OJO: sin rdf:type; solo nombre humano (rdfs:label).
    lbl(EX[pid], nombre)

# --- Niveles de estudios (4 individuos) -----------------------------
NIVELES = {
    "nivel_pregrado":  "Pregrado",
    "nivel_posgrado":  "Posgrado",
    "nivel_maestria":  "Maestría",
    "nivel_doctorado": "Doctorado",
}
for nid, nombre in NIVELES.items():
    g.add((EX[nid], RDF.type, NivelEstudios))
    lbl(EX[nid], nombre)

# --- Áreas de conocimiento ------------------------------------------
# La subclase Ingenieria se puebla con >=4 individuos, uno por cada
# ingeniería. Cada uno se declara como Ingenieria: el razonador
# inferirá que también son AreaConocimiento -> "caso 1b" (jerarquía
# de clases).
INGENIERIA_INDIVIDUOS = {
    "area_ingenieria_sistemas":    "Ingeniería de Sistemas",
    "area_ingenieria_civil":       "Ingeniería Civil",
    "area_ingenieria_electronica": "Ingeniería Electrónica",
    "area_ingenieria_industrial":  "Ingeniería Industrial",
    "area_ingenieria_mecanica":    "Ingeniería Mecánica",
}
for aid, nombre in INGENIERIA_INDIVIDUOS.items():
    g.add((EX[aid], RDF.type, Ingenieria))
    lbl(EX[aid], nombre)
# El resto de áreas son individuos directos de AreaConocimiento.
# [C9] Se agregan Arquitectura y Humanidades: son las áreas de dos de los
# estudiantes del catálogo (antes no existían en la ontología).
AREAS = {
    "area_comun":       "Todas las áreas comunes",
    "area_matematicas": "Matemáticas y Ciencias Naturales",
    "area_arquitectura":"Arquitectura",
    "area_humanidades": "Humanidades",
}
for aid, nombre in AREAS.items():
    g.add((EX[aid], RDF.type, AreaConocimiento))
    lbl(EX[aid], nombre)

# --- Convenios (4+ individuos) --------------------------------------
CONVENIOS = {
    "convenio_unal_fub":               "Convenio bilateral UNAL - Freie Universität Berlin",
    "convenio_unal_montpellier":       "Convenio bilateral UNAL - Université de Montpellier",
    "convenio_unal_unam":              "Convenio bilateral UNAL - UNAM",
    "convenio_unal_escuela_colombiana":"Convenio SÍGUEME UNAL - Escuela Colombiana de Ingeniería",
}
for cid, titulo in CONVENIOS.items():
    g.add((EX[cid], RDF.type, Convenio))
    g.add((EX[cid], DCTERMS.title, Literal(titulo, lang="es")))  # convenio = documento: dct:title

# --- Becas (4 individuos) -------------------------------------------
BECAS = {
    "beca_fub":             ("Beca FUB Summer Semester",
                             "Apoyo mensual de 450 EUR, seguro médico, reembolso administrativo y exención de matrícula."),
    "beca_santander":       ("Beca Santander Movilidad",
                             "Apoyo económico parcial para gastos de movilidad internacional."),
    "beca_alianza_pacifico":("Beca Plataforma Alianza del Pacífico",
                             "Beca de intercambio para estudiantes de los países de la Alianza del Pacífico."),
    "beca_icetex":          ("Crédito-beca ICETEX",
                             "Financiación parcial de movilidad para estudiantes colombianos."),
}
for bid, (titulo, desc) in BECAS.items():
    g.add((EX[bid], RDF.type, Beca))
    g.add((EX[bid], DCTERMS.title, Literal(titulo, lang="es")))       # beca = programa/documento
    g.add((EX[bid], DCTERMS.description, Literal(desc, lang="es")))

# --- Estudiantes (4 individuos, Estudiante ⊑ foaf:Person) -----------
# [C9] Cada estudiante trae ahora su PERFIL completo: área, nivel,
# promedio e idiomas. El sistema experto y la lógica difusa lo leen del
# grafo; ya no se escribe a mano en Python.
ESTUDIANTES = {
    "est_juan_sistemas": dict(
        nombre="Juan Valencia (Ingeniería de Sistemas)", area="area_ingenieria_sistemas",
        nivel="nivel_pregrado", promedio="4.3",
        idiomas=["Español nativo", "Inglés B2", "Francés A2"]),
    "est_maria_matematicas": dict(
        nombre="María Gómez (Matemáticas)", area="area_matematicas",
        nivel="nivel_pregrado", promedio="4.7",
        idiomas=["Español nativo", "Inglés C1"]),
    "est_carlos_arquitectura": dict(
        nombre="Carlos Ruiz (Arquitectura)", area="area_arquitectura",
        nivel="nivel_maestria", promedio="3.6",
        idiomas=["Español nativo", "Inglés A2"]),
    "est_ana_humanidades": dict(
        nombre="Ana Torres (Humanidades)", area="area_humanidades",
        nivel="nivel_pregrado", promedio="3.1",
        idiomas=["Español nativo", "Portugués B1"]),
}
for eid, d in ESTUDIANTES.items():
    est = EX[eid]
    g.add((est, RDF.type, Estudiante))
    g.add((est, FOAF.name, Literal(d["nombre"])))   # Estudiante ⊑ foaf:Person: su nombre va en foaf:name
    g.add((est, EX.estudiaArea, EX[d["area"]]))
    g.add((est, EX.cursaNivel, EX[d["nivel"]]))
    g.add((est, EX.tienePromedio, Literal(d["promedio"], datatype=XSD.decimal)))
    for idioma in d["idiomas"]:
        g.add((est, EX.dominaIdioma, Literal(idioma, datatype=XSD.string)))


# ------------------------------------------------------------------
# 6. UNIVERSIDADES E INDIVIDUOS DE OFERTA (datos reales del Excel DRE/UNAL)
# ------------------------------------------------------------------
# Universidad de ORIGEN: se deja SIN rdf:type para demostrar el
# "caso 2" por DOMINIO: uni_unal ubicadaEn pais_colombia y como el
# domain de ubicadaEn es Universidad, el razonador infiere que
# uni_unal es Universidad.
g.add((EX["uni_unal"], FOAF.name, Literal("Universidad Nacional de Colombia")))
g.add((EX["uni_unal"], EX.ubicadaEn, EX["pais_colombia"]))

# Cada oferta se describe con un diccionario. Datos tomados del archivo
# "Ofertas Específicas - Internacionales - 2027-1.xlsx" (hojas 1 y 2).
#   tipo      : "int" (OfertaInternacional) | "nac" (OfertaNacional)
#   uni_name  : nombre humano de la universidad de destino
#   pais      : id del país (ex:pais_*)
#   cierre    : fecha ISO 'YYYY-MM-DD' o None
#   niveles   : lista de ids ex:nivel_*
#   areas     : lista de ids de área para tieneArea
#   restringe : id de área para restringeArea, o None
#   cupos     : int o None (None = "sujeto a aprobación", no se declara)
#   idioma    : str o None
#   beca      : True/False/None
#   becas     : [C2] lista de ids ex:beca_* (tieneBeca)
#   convenio  : [C2] id ex:convenio_* (amparadaPorConvenio) o None
#   desc      : nota relevante (dct:description) o None
OFERTAS = {
 # ---------------- Ofertas internacionales ----------------
 "oferta_montpellier_2027_1": dict(
     tipo="int", titulo="Intercambio en Université de Montpellier Paul Valéry",
     uni_name="Université de Montpellier Paul Valéry", pais="pais_francia",
     cierre="2026-09-29", niveles=["nivel_pregrado"],
     areas=["area_comun"], restringe=None, cupos=4, idioma="Francés B1",
     beca=None, becas=[], convenio="convenio_unal_montpellier", desc=None),

 "oferta_unam_2027_1": dict(
     tipo="int", titulo="Intercambio en la UNAM",
     uni_name="Universidad Nacional Autónoma de México (UNAM)", pais="pais_mexico",
     cierre="2026-08-20", niveles=["nivel_pregrado"],
     areas=["area_comun"], restringe=None, cupos=10, idioma="Español (B2 para no hispanohablantes)",
     beca=False, becas=[], convenio="convenio_unal_unam",
     desc="Requiere GPA 8/10 (escala UNAM) y avance mínimo del 40% del plan de estudios."),

 "oferta_freie_berlin_2027_1": dict(
     tipo="int", titulo="Intercambio en Freie Universität Berlin",
     uni_name="Freie Universität Berlin", pais="pais_alemania",
     cierre="2026-09-30", niveles=["nivel_pregrado"],
     areas=["area_comun"], restringe=None, cupos=4, idioma="Inglés B2 / Alemán",
     beca=True, becas=["beca_fub"], convenio="convenio_unal_fub",
     desc="Beca FUB para 2 estudiantes: 450 EUR/mes, seguro médico y exención de matrícula."),

 "oferta_stavanger_2027_1": dict(
     tipo="int", titulo="Intercambio en University of Stavanger",
     uni_name="University of Stavanger", pais="pais_noruega",
     cierre="2026-09-09", niveles=["nivel_pregrado"],
     areas=["area_matematicas"], restringe="area_matematicas", cupos=None, idioma=None,
     beca=None, becas=[], convenio=None,
     desc="Oferta exclusiva para Matemáticas y Ciencias Naturales."),

 # ---------------- Ofertas nacionales ----------------
 # [C10] La Escuela Colombiana lista EXPLÍCITAMENTE Ingeniería de Sistemas
 # (así lo dice su descripción): se modela con dos tieneArea.
 "oferta_escuela_colombiana_2027_1": dict(
     tipo="nac", titulo="Intercambio SÍGUEME en Escuela Colombiana de Ingeniería Julio Garavito",
     uni_name="Escuela Colombiana de Ingeniería Julio Garavito", pais="pais_colombia",
     cierre="2026-10-11", niveles=["nivel_pregrado", "nivel_posgrado", "nivel_maestria", "nivel_doctorado"],
     areas=["area_comun", "area_ingenieria_sistemas"], restringe=None, cupos=5, idioma=None,
     beca=True, becas=["beca_icetex"], convenio="convenio_unal_escuela_colombiana",
     desc="Oferta que incluye explícitamente Ingeniería de Sistemas. Requiere el 50% de los créditos aprobados."),

 "oferta_eafit_2027_1": dict(
     tipo="nac", titulo="Intercambio en Universidad EAFIT",
     uni_name="Universidad EAFIT", pais="pais_colombia",
     cierre="2026-09-18", niveles=["nivel_pregrado"],
     areas=["area_comun"], restringe=None, cupos=9, idioma=None,
     beca=None, becas=[], convenio=None, desc=None),

 "oferta_ufps_2027_1": dict(
     tipo="nac", titulo="Intercambio en Universidad Francisco de Paula Santander (Ocaña)",
     uni_name="Universidad Francisco de Paula Santander Ocaña", pais="pais_colombia",
     cierre="2026-10-25", niveles=["nivel_pregrado"],
     areas=["area_comun"], restringe=None, cupos=10, idioma=None,
     beca=None, becas=[], convenio=None, desc=None),

 "oferta_tolima_2027_1": dict(
     tipo="nac", titulo="Intercambio en Universidad del Tolima",
     uni_name="Universidad del Tolima", pais="pais_colombia",
     cierre="2026-09-18", niveles=["nivel_pregrado"],
     areas=["area_comun"], restringe=None, cupos=None, idioma=None,
     beca=None, becas=[], convenio=None, desc=None),
}

# Tipo de oferta -> clase RDF
CLASE_OFERTA = {"int": OfertaInternacional, "nac": OfertaNacional}

for oid, d in OFERTAS.items():
    oferta = EX[oid]
    # --- tipo de oferta (explícito) -> "caso 1a": se infiere ex:Oferta ---
    g.add((oferta, RDF.type, CLASE_OFERTA[d["tipo"]]))
    g.add((oferta, DCTERMS.title, Literal(d["titulo"], lang="es")))   # oferta = convocatoria/documento
    if d["desc"]:
        g.add((oferta, DCTERMS.description, Literal(d["desc"], lang="es")))

    # --- universidad de destino (individuo derivado del id de la oferta) ---
    uni_id = "uni_" + oid[len("oferta_"):]          # p.ej. uni_montpellier_2027_1
    uni = EX[uni_id]
    g.add((uni, RDF.type, UniversidadDestino))       # UniversidadDestino ⊑ Universidad
    g.add((uni, FOAF.name, Literal(d["uni_name"])))  # universidad = organización/agente: foaf:name
    g.add((uni, EX.ubicadaEn, EX[d["pais"]]))        # domain Universidad / range Pais

    # --- ofrecidaPorUniversidadDestino (subpropiedad) -> "caso 3" ---
    g.add((oferta, EX.ofrecidaPorUniversidadDestino, uni))

    # --- país de la oferta (range Pais) -> "caso 2" (rango) ---
    g.add((oferta, EX.ofreceEnPais, EX[d["pais"]]))

    # --- área(s), restricción y nivel ---
    for area in d["areas"]:
        g.add((oferta, EX.tieneArea, EX[area]))
    if d["restringe"]:
        g.add((oferta, EX.restringeArea, EX[d["restringe"]]))
    for niv in d["niveles"]:
        g.add((oferta, EX.dirigidaANivel, EX[niv]))

    # --- [C2] becas y convenio asociados a la oferta ---
    for beca in d["becas"]:
        g.add((oferta, EX.tieneBeca, EX[beca]))
    if d["convenio"]:
        g.add((oferta, EX.amparadaPorConvenio, EX[d["convenio"]]))

    # --- literales tipados (XSD) ---
    if d["cupos"] is not None:
        g.add((oferta, EX.tieneCupos, Literal(d["cupos"], datatype=XSD.integer)))
    if d["cierre"] is not None:
        g.add((oferta, EX.fechaCierre, Literal(d["cierre"], datatype=XSD.date)))
    if d["idioma"] is not None:
        g.add((oferta, EX.requiereIdioma, Literal(d["idioma"], datatype=XSD.string)))
    if d["beca"] is not None:
        g.add((oferta, EX.ofreceBeca, Literal(d["beca"], datatype=XSD.boolean)))


# ------------------------------------------------------------------
# 7. SERIALIZACIÓN A TURTLE (archivo canónico)
# ------------------------------------------------------------------
import os
# Ruta de salida: junto al script si se ejecuta como archivo .py; en el
# directorio de trabajo si se ejecuta dentro de una celda de Colab/Jupyter
# (donde la variable __file__ no existe).
BASE_DIR = os.path.dirname(os.path.abspath(__file__)) if "__file__" in globals() else os.getcwd()
OUT = os.path.join(BASE_DIR, "ontologia.ttl")
g.serialize(destination=OUT, format="turtle")

print(f"[OK] Ontología serializada en: {OUT}")
print(f"[OK] Total de triples asertados (antes de razonar): {len(g)}")

if __name__ == "__main__":
    # Pequeño resumen de verificación de mínimos.
    n_clases = len(set(g.subjects(RDF.type, RDFS.Class)))
    n_sub    = len(list(g.triples((None, RDFS.subClassOf, None))))
    n_props  = len(set(g.subjects(RDF.type, RDF.Property)))
    n_subp   = len(list(g.triples((None, RDFS.subPropertyOf, None))))
    print("-" * 55)
    print(f"  Clases declaradas        : {n_clases}  (min 10)")
    print(f"  Jerarquías subClassOf    : {n_sub}  (min 5)")
    print(f"  Propiedades declaradas   : {n_props}  (min 10)")
    print(f"  Jerarquías subPropertyOf : {n_subp}  (min 1)")
```

## Bloque 3 — `razonamiento.py`

Aplica el razonador OWL-RL (RDFS) y evidencia los casos 1, 2 y 3, la comparación antes/después y los hechos nuevos relevantes inferidos (tipos + propiedades del contrato).

```python
# -*- coding: utf-8 -*-
"""
razonamiento.py
==================================================================
Carga la ontología canónica `ontologia.ttl`, aplica el razonador
OWL-RL con `DeductiveClosure(RDFS_Semantics)` y EVIDENCIA los
requisitos de razonamiento del enunciado (§2.3):

  - Caso 1 (>=2): generación de hechos por jerarquía de clases.
  - Caso 2      : membresía de clase por el DOMINIO o el RANGO de
                  una propiedad.
  - Caso 3      : nuevos hechos por rdfs:subPropertyOf.
  - Comparación del grafo ANTES y DESPUÉS del razonamiento.
  - Evidencia de al menos 4 hechos nuevos inferidos.

IMPORTANTE (arquitectura §6.1): el razonador expande el grafo SOLO
en memoria; NUNCA reescribe el archivo `ontologia.ttl` en disco.
==================================================================
"""

import os
from rdflib import Graph, Namespace
from rdflib.namespace import RDF, RDFS
from owlrl import DeductiveClosure, RDFS_Semantics

EX = Namespace("http://intercambios.unal.edu.co/")
# Ruta del .ttl: junto al script si se ejecuta como archivo .py; en el
# directorio de trabajo si se ejecuta dentro de una celda de Colab/Jupyter
# (donde la variable __file__ no existe).
BASE_DIR = os.path.dirname(os.path.abspath(__file__)) if "__file__" in globals() else os.getcwd()
TTL = os.path.join(BASE_DIR, "ontologia.ttl")


def cargar_grafo():
    """Carga la ontología canónica desde el .ttl (fuente única de datos)."""
    g = Graph()
    g.parse(TTL, format="turtle")
    return g


def qn(uri):
    """Devuelve el nombre corto (prefijo:local) de un URI para imprimir."""
    s = str(uri)
    for pref, ns in (("ex:", str(EX)), ("rdfs:", str(RDFS)), ("rdf:", str(RDF))):
        if s.startswith(ns):
            return pref + s[len(ns):]
    return s


def existe(g, s, p, o):
    """True si el triple (s, p, o) está presente en el grafo g."""
    return (s, p, o) in g


def main():
    # ---------- 1. Grafo ANTES de razonar ----------
    g = cargar_grafo()
    triples_antes = len(g)
    # Snapshot del conjunto de triples para poder aislar lo NUEVO.
    antes = set(g)

    print("=" * 62)
    print("  RAZONAMIENTO RDFS  —  Ontología de intercambios (UNAL)")
    print("=" * 62)
    print(f"Triples ANTES del razonamiento : {triples_antes}")

    # ---------- 2. Aplicar el razonador OWL-RL ----------
    # DeductiveClosure con la semántica RDFS: aplica las reglas de
    # subClassOf, subPropertyOf, domain y range sobre el grafo.
    DeductiveClosure(RDFS_Semantics).expand(g)

    triples_despues = len(g)
    print(f"Triples DESPUÉS del razonamiento: {triples_despues}")
    print(f"NUEVOS triples inferidos        : {triples_despues - triples_antes}")
    print()

    # Conjunto de triples inferidos (los que no estaban antes).
    nuevos = set(g) - antes

    # ---------- 3. Verificación de los casos requeridos ----------
    # Cada 'caso' es (descripcion, triple_inferido_esperado).
    casos = [
        # ---- CASO 1: jerarquía de clases (>=2) ----
        ("CASO 1a — Jerarquía de clases: oferta declarada como "
         "OfertaInternacional resulta ser Oferta",
         (EX.oferta_montpellier_2027_1, RDF.type, EX.Oferta)),

        ("CASO 1b — Jerarquía de clases: el área declarada como "
         "Ingenieria resulta ser AreaConocimiento",
         (EX.area_ingenieria_sistemas, RDF.type, EX.AreaConocimiento)),

        # Extra de jerarquía: UniversidadDestino ⊑ Universidad
        ("CASO 1c — Jerarquía de clases: una UniversidadDestino "
         "resulta ser Universidad",
         (EX.uni_montpellier_2027_1, RDF.type, EX.Universidad)),

        # ---- CASO 2: dominio / rango ----
        ("CASO 2 (RANGO) — pais_noruega (sin tipo declarado) resulta ser "
         "Pais por el range de ofreceEnPais",
         (EX.pais_noruega, RDF.type, EX.Pais)),

        ("CASO 2 (DOMINIO) — uni_unal (sin tipo declarado) resulta ser "
         "Universidad por el domain de ubicadaEn",
         (EX.uni_unal, RDF.type, EX.Universidad)),

        # ---- CASO 3: subPropertyOf ----
        ("CASO 3 — subPropertyOf: ofrecidaPorUniversidadDestino implica "
         "ofrecidaPor",
         (EX.oferta_montpellier_2027_1, EX.ofrecidaPor, EX.uni_montpellier_2027_1)),
    ]

    print("-" * 62)
    print("  VERIFICACIÓN DE LOS CASOS DE GENERACIÓN DE CONOCIMIENTO")
    print("-" * 62)
    todos_ok = True
    for desc, triple in casos:
        s, p, o = triple
        ok = existe(g, s, p, o)
        era_nuevo = triple in nuevos
        estado = "OK" if ok else "FALLA"
        marca = "  [INFERIDO]" if era_nuevo else "  [ya existía]"
        if not ok:
            todos_ok = False
            marca = ""
        print(f"[{estado}] {desc}")
        print(f"        ({qn(s)}  {qn(p)}  {qn(o)}){marca}")
    print()

    # ---------- 4. Muestra de los hechos nuevos RELEVANTES ----------
    # El cierre RDFS produce dos tipos de triples nuevos:
    #   (a) útiles del dominio -> membresías de clase (rdf:type hacia una
    #       clase ex:) y las propiedades del contrato §4.2 que se infieren
    #       (p. ej. ofrecidaPor por subPropertyOf);
    #   (b) ruido de la semántica RDFS -> rdf:type rdfs:Resource, subClassOf
    #       reflexivo, variantes de literales, declaraciones de propiedad...
    # Aquí mostramos SOLO (a): todo lo que el traductor mandaría al experto.
    # [C2] Se incluyen las 6 propiedades nuevas del contrato.
    PROPS_CONTRATO = {
        EX.ofrecidaPor, EX.ofrecidaPorUniversidadDestino, EX.ofreceEnPais,
        EX.tieneArea, EX.dirigidaANivel, EX.restringeArea, EX.ubicadaEn,
        EX.tieneCupos, EX.fechaCierre, EX.requiereIdioma, EX.ofreceBeca,
        EX.tieneBeca, EX.amparadaPorConvenio, EX.estudiaArea, EX.cursaNivel,
        EX.tienePromedio, EX.dominaIdioma,
    }
    relevantes = [
        (s, p, o) for (s, p, o) in nuevos
        if (p == RDF.type and str(o).startswith(str(EX))) or p in PROPS_CONTRATO
    ]
    print("-" * 62)
    print("  HECHOS NUEVOS RELEVANTES INFERIDOS (tipos + propiedades §4.2)")
    print(f"  Total: {len(relevantes)}  (el enunciado exige >= 4)")
    print("-" * 62)
    for s, p, o in sorted(relevantes, key=lambda t: (str(t[1]), str(t[0]))):
        print(f"   {qn(s):38}  {qn(p):16}  {qn(o)}")

    print()
    print("=" * 62)
    resumen = "TODOS LOS CASOS SE CUMPLEN" if todos_ok else "HAY CASOS SIN CUMPLIR"
    print(f"  RESULTADO: {resumen}")
    print(f"  Hechos nuevos totales: {triples_despues - triples_antes} | "
          f"Hechos de dominio inferidos: {len(relevantes)}")
    print("=" * 62)


if __name__ == "__main__":
    main()
```

## Lógica difusa (Scikit-Fuzzy)

Calcula la **idoneidad** (0–100) de cada par (estudiante, oferta) a partir de `promedio`, `nivel_idioma` y `cupos`.

**[C4/C9]** La demo ya no usa datos escritos a mano: carga y razona la ontología y lee de ella las ofertas (`ofertas_desde_grafo`) y los perfiles de los estudiantes (`perfil_desde_grafo`). Se agregan `interpretar_dominio` y `cumple_idioma`, que usa la regla r17 del sistema experto.

```python
!pip install scikit-fuzzy
```

```python
import numpy as np
import skfuzzy as fuzz
from skfuzzy import control as ctrl

# ---------------------------------------------------------------------------
# 1. UNIVERSOS DEL DISCURSO
# ---------------------------------------------------------------------------
promedio = ctrl.Antecedent(np.arange(0, 5.01, 0.01), 'promedio')
nivel_idioma = ctrl.Antecedent(np.arange(0, 101, 1), 'nivel_idioma')
cupos = ctrl.Antecedent(np.arange(0, 10.01, 0.05), 'cupos')
idoneidad = ctrl.Consequent(np.arange(0, 101, 1), 'idoneidad')

# ---------------------------------------------------------------------------
# 2. FUNCIONES DE PERTENENCIA
# ---------------------------------------------------------------------------
# Promedio (escala UNAL 0-5): TRAPEZOIDALES en los extremos + triangular al centro.
#   < 3.0 es claramente bajo (3.0 es la nota mínima aprobatoria)
#   > 4.5 es claramente alto (la mayoría de convocatorias piden >= 3.5 o 4.0)
promedio['bajo'] = fuzz.trapmf(promedio.universe, [0, 0, 3.0, 3.6])
promedio['medio'] = fuzz.trimf(promedio.universe, [3.2, 3.8, 4.4])
promedio['alto'] = fuzz.trapmf(promedio.universe, [4.0, 4.5, 5.0, 5.0])

# Nivel de idioma (0-100): TRIANGULARES. Ver `mcer_a_puntaje` para convertir A1..C2.
nivel_idioma['basico'] = fuzz.trimf(nivel_idioma.universe, [0, 0, 45])
nivel_idioma['intermedio'] = fuzz.trimf(nivel_idioma.universe, [30, 55, 80])
nivel_idioma['avanzado'] = fuzz.trimf(nivel_idioma.universe, [65, 100, 100])

# Cupos disponibles (0-10, en el Excel van de 1 a 10): GAUSSIANAS.
#   Pocos cupos => oferta más competida.
cupos['pocos'] = fuzz.gaussmf(cupos.universe, 1.0, 1.5)
cupos['moderados'] = fuzz.gaussmf(cupos.universe, 5.0, 1.5)
cupos['muchos'] = fuzz.gaussmf(cupos.universe, 10.0, 2.0)

# Salida: idoneidad (0-100), TRIANGULARES.
idoneidad['baja'] = fuzz.trimf(idoneidad.universe, [0, 0, 50])
idoneidad['media'] = fuzz.trimf(idoneidad.universe, [25, 50, 75])
idoneidad['alta'] = fuzz.trimf(idoneidad.universe, [50, 100, 100])

# ---------------------------------------------------------------------------
# 3. MODIFICADORES (hedges)
# ---------------------------------------------------------------------------
def muy(mf):
    """Concentración: mu_muy(x) = mu(x)^2. Hace el conjunto más estricto."""
    return np.power(mf, 2)


def ligeramente(mf):
    """Dilatación: mu_lig(x) = mu(x)^0.5. Hace el conjunto más permisivo."""
    return np.power(mf, 0.5)


# Se agregan como términos nuevos para poder usarlos DENTRO de las reglas.
promedio['muy_alto'] = muy(promedio['alto'].mf)
cupos['ligeramente_pocos'] = ligeramente(cupos['pocos'].mf)

# ---------------------------------------------------------------------------
# 4. BASE DE REGLAS (10 reglas: AND = &, OR = |, NOT = ~)
# ---------------------------------------------------------------------------
reglas = [
    # R1  AND: buen perfil completo
    ctrl.Rule(promedio['alto'] & nivel_idioma['avanzado'], idoneidad['alta'], label='R1'),
    # R2  AND + modificador "muy": un promedio excelente compensa la alta competencia
    ctrl.Rule(promedio['muy_alto'] & cupos['pocos'], idoneidad['alta'], label='R2'),
    # R3  OR: cualquiera de las dos debilidades hunde la postulación
    ctrl.Rule(promedio['bajo'] | nivel_idioma['basico'], idoneidad['baja'], label='R3'),
    # R4  AND: promedio regular en una oferta muy competida
    ctrl.Rule(promedio['medio'] & cupos['pocos'], idoneidad['baja'], label='R4'),
    # R5  AND + NOT: promedio regular pero la oferta NO es muy competida
    ctrl.Rule(promedio['medio'] & ~cupos['pocos'], idoneidad['media'], label='R5'),
    # R6  AND: perfil intermedio
    ctrl.Rule(promedio['medio'] & nivel_idioma['intermedio'], idoneidad['media'], label='R6'),
    # R7  AND (3 antecedentes): buen promedio, idioma suficiente y oferta amplia
    ctrl.Rule(promedio['alto'] & nivel_idioma['intermedio'] & cupos['muchos'],
              idoneidad['alta'], label='R7'),
    # R8  AND + modificador "ligeramente"
    ctrl.Rule(nivel_idioma['avanzado'] & cupos['ligeramente_pocos'], idoneidad['media'], label='R8'),
    # R9  NOT + AND: idioma al menos intermedio y muchos cupos
    ctrl.Rule(~nivel_idioma['basico'] & cupos['muchos'], idoneidad['alta'], label='R9'),
    # R10 OR: situaciones neutras
    ctrl.Rule(promedio['medio'] | cupos['moderados'], idoneidad['media'], label='R10'),
]

# ---------------------------------------------------------------------------
# 5. SISTEMA DE CONTROL Y DEFUZZIFICACIÓN
# ---------------------------------------------------------------------------
idoneidad.defuzzify_method = 'centroid'
sistema = ctrl.ControlSystem(reglas)

# Umbrales derivados de la salida para pasar la categoría al sistema experto
_CATEGORIAS = ['baja', 'media', 'alta']


def categoria_de(puntaje):
    """Devuelve el término de salida con mayor grado de pertenencia para `puntaje`."""
    grados = {c: fuzz.interp_membership(idoneidad.universe, idoneidad[c].mf, puntaje)
              for c in _CATEGORIAS}
    return max(grados, key=grados.get), grados


def evaluar(prom, idioma, n_cupos, metodo='centroid'):
    """
    Ejecuta la inferencia Mamdani para un par (estudiante, oferta).

    Parámetros
    ----------
    prom    : float  promedio del estudiante (0-5)
    idioma  : float  nivel del estudiante en el idioma requerido (0-100)
    n_cupos : float  cupos disponibles de la oferta (se recorta a 0-10)
    metodo  : str    'centroid' (por defecto), 'bisector', 'mom', 'som', 'lom'

    Retorna
    -------
    dict con 'puntaje' (float), 'categoria' (str) y 'grados' (pertenencias).
    """
    idoneidad.defuzzify_method = metodo
    try:
        sim = ctrl.ControlSystemSimulation(sistema)
        sim.input['promedio'] = float(np.clip(prom, 0, 5))
        sim.input['nivel_idioma'] = float(np.clip(idioma, 0, 100))
        sim.input['cupos'] = float(np.clip(n_cupos, 0, 10))
        sim.compute()
    finally:
        # Se restaura SIEMPRE el método por defecto, aunque ocurra un error.
        idoneidad.defuzzify_method = 'centroid'
    puntaje = float(sim.output['idoneidad'])
    cat, grados = categoria_de(puntaje)
    return {'puntaje': round(puntaje, 2), 'categoria': cat,
            'grados': {k: round(float(v), 3) for k, v in grados.items()}, 'sim': sim}


# ---------------------------------------------------------------------------
# 6. UTILIDADES DE INTEGRACIÓN (lectura del grafo + interpretación de idiomas)
# ---------------------------------------------------------------------------
import re
import unicodedata

# Escala MCER -> 0-100 (misma escala del universo de nivel_idioma)
_MCER = {'A1': 10, 'A2': 25, 'B1': 45, 'B2': 65, 'C1': 85, 'C2': 100, 'NATIVO': 100}

# Palabras con las que se reconoce cada idioma dentro del texto de la ontología
_IDIOMAS = {'espanol': 'espanol', 'ingles': 'ingles', 'frances': 'frances',
            'aleman': 'aleman', 'portugues': 'portugues', 'italiano': 'italiano',
            'coreano': 'coreano', 'noruego': 'noruego'}

# Si la ontología no trae cupos (p. ej. "sujeto a aprobación"), se asume un valor
# neutro: 5 cupos = centro del conjunto "moderados". Ni premia ni castiga la oferta.
CUPOS_DESCONOCIDOS = 5


def _normalizar(texto):
    """Minúsculas y sin tildes: 'Francés B1' -> 'frances b1'."""
    t = unicodedata.normalize('NFD', str(texto).lower())
    return ''.join(c for c in t if unicodedata.category(c) != 'Mn')


def mcer_a_puntaje(nivel):
    """Convierte un nivel MCER ('B2', 'C1', 'nativo'...) a la escala 0-100."""
    if nivel is None:
        return 0
    return _MCER.get(str(nivel).strip().upper(), 0)


def idiomas_requeridos(texto_idioma, tipo):
    """
    Interpreta el texto libre de requisito de idioma de una oferta.

    'Francés B1'                            -> [('frances', 'B1')]
    'Inglés B2 / Alemán'                    -> [('ingles', 'B2'), ('aleman', None)]
    'Español (B2 para no hispanohablantes)' -> [('espanol', 'B2')]
    None en oferta nacional                 -> [('espanol', None)]
    None en oferta internacional            -> [('ingles', None)]   (idioma académico por defecto)
    """
    if not texto_idioma:
        return [('espanol', None)] if tipo == 'nac' else [('ingles', None)]
    resultado = []
    for parte in re.split(r'[/,;]| o ', _normalizar(texto_idioma)):
        idioma = next((v for k, v in _IDIOMAS.items() if k in parte), None)
        if idioma:
            m = re.search(r'\b([abc][12])\b', parte)
            resultado.append((idioma, m.group(1).upper() if m else None))
    return resultado or [('ingles', None)]


def interpretar_dominio(texto):
    """
    [C3/C9] Interpreta un valor de ex:dominaIdioma del estudiante.
    'Inglés B2' -> ('ingles', 'B2')     'Español nativo' -> ('espanol', 'NATIVO')
    """
    t = _normalizar(texto)
    idioma = next((v for k, v in _IDIOMAS.items() if k in t), None)
    if 'nativo' in t:
        return idioma, 'NATIVO'
    m = re.search(r'\b([abc][12])\b', t)
    return idioma, (m.group(1).upper() if m else None)


def cumple_idioma(requisito, dominio):
    """
    [C3] True si UN idioma que domina el estudiante (texto de ex:dominaIdioma)
    satisface el requisito de la oferta (texto de ex:requiereIdioma).
    Si la oferta acepta varios idiomas ('Inglés B2 / Alemán') basta con uno.
    Un requisito sin nivel explícito ('Alemán') exige al menos A1.
    La usa la regla r17 del sistema experto dentro de un TEST(...).
    """
    idioma_est, nivel_est = interpretar_dominio(dominio)
    for idioma_req, nivel_req in idiomas_requeridos(requisito, 'int'):
        if idioma_req == idioma_est and mcer_a_puntaje(nivel_est) >= mcer_a_puntaje(nivel_req or 'A1'):
            return True
    return False


def nivel_para_oferta(estudiante, oferta):
    """
    Nivel (0-100) del estudiante en el idioma que exige la oferta.
    Si la oferta acepta varios idiomas, se toma el que mejor domine el estudiante.
    """
    niveles_est = {_normalizar(k): v for k, v in estudiante.get('idiomas', {}).items()}
    opciones = idiomas_requeridos(oferta.get('idioma'), oferta.get('tipo'))
    return max(mcer_a_puntaje(niveles_est.get(idioma)) for idioma, _ in opciones)


def evaluar_oferta(estudiante, oferta_id, oferta):
    """Evalúa una oferta y devuelve el resultado con el detalle de las entradas usadas."""
    cupos_oferta = oferta.get('cupos')
    n_cupos = CUPOS_DESCONOCIDOS if cupos_oferta is None else cupos_oferta
    nivel = nivel_para_oferta(estudiante, oferta)
    r = evaluar(estudiante['promedio'], nivel, n_cupos)
    return {'oferta': oferta_id, 'puntaje': r['puntaje'], 'categoria': r['categoria'],
            'entradas': {'promedio': estudiante['promedio'], 'nivel_idioma': nivel,
                         'cupos': n_cupos, 'cupos_estimados': cupos_oferta is None}}


def resultados_para_experto(estudiante, ofertas):
    """
    Evalúa todas las ofertas para un estudiante.

    estudiante : {'promedio': 4.3, 'idiomas': {'ingles': 'B2', 'espanol': 'NATIVO'}}
                 (lo devuelve `perfil_desde_grafo`)
    ofertas    : {id_oferta: {'tipo': 'int'|'nac', 'cupos': int|None, 'idioma': str|None}, ...}
                 (lo devuelve `ofertas_desde_grafo`)

    Retorna una lista de dicts {'oferta', 'puntaje', 'categoria'}
    que se declaran en Experta como ResultadoDifuso(**d).
    """
    return [{k: v for k, v in evaluar_oferta(estudiante, oid, o).items() if k != 'entradas'}
            for oid, o in ofertas.items()]


NS_ONTOLOGIA = 'http://intercambios.unal.edu.co/'


def ofertas_desde_grafo(g, ns=NS_ONTOLOGIA):
    """
    Lee del grafo RDF ya razonado los datos que necesita la lógica difusa,
    SIN nombres ni valores fijos: recorre todos los individuos de tipo ex:Oferta.

    Usa los predicados del contrato: ex:tieneCupos y ex:requiereIdioma,
    y la clase ex:OfertaNacional para saber si la oferta es nacional.
    """
    from rdflib import Namespace, RDF
    EX_ = Namespace(ns)
    ofertas = {}
    for s in set(g.subjects(RDF.type, EX_.Oferta)):
        oid = str(s).replace(ns, '')
        cupos_lit = g.value(s, EX_.tieneCupos)
        idioma = g.value(s, EX_.requiereIdioma)
        ofertas[oid] = {
            'tipo': 'nac' if (s, RDF.type, EX_.OfertaNacional) in g else 'int',
            'cupos': int(cupos_lit) if cupos_lit is not None else None,
            'idioma': str(idioma) if idioma is not None else None,
        }
    if not ofertas:
        # [C-T4] Antes devolvía {} sin avisar si el grafo no estaba razonado.
        print('[AVISO] No se encontró ningún ex:Oferta en el grafo. '
              '¿Se aplicó DeductiveClosure(RDFS_Semantics) antes de llamar esta función?')
    return dict(sorted(ofertas.items()))


def perfil_desde_grafo(g, estudiante_id, ns=NS_ONTOLOGIA):
    """
    [C9] Lee del grafo el perfil que la lógica difusa necesita de un estudiante
    (ex:tienePromedio y ex:dominaIdioma). Ya no se escribe a mano en Python.

    Retorna {'promedio': float, 'idiomas': {'ingles': 'B2', 'espanol': 'NATIVO', ...}}
    """
    from rdflib import Namespace
    EX_ = Namespace(ns)
    est = EX_[estudiante_id]
    promedio = g.value(est, EX_.tienePromedio)
    if promedio is None:
        raise ValueError(f"El estudiante '{estudiante_id}' no tiene ex:tienePromedio en la ontología")
    idiomas = {}
    for dom in g.objects(est, EX_.dominaIdioma):
        idioma, nivel = interpretar_dominio(dom)
        if idioma:
            idiomas[idioma] = nivel
    return {'promedio': float(promedio), 'idiomas': idiomas}


# ---------------------------------------------------------------------------
# 7. GRÁFICAS PARA EL DOCUMENTO DE DISEÑO
# ---------------------------------------------------------------------------
def graficar_todo(carpeta='.', ejemplo=(4.3, 70, 3)):
    """Guarda las gráficas de pertenencia (con modificadores) y la salida defuzzificada."""
    import os
    import matplotlib.pyplot as plt

    def _plot(var, titulo, archivo, destacar=()):
        fig, ax = plt.subplots(figsize=(8, 3.8))
        for nombre, term in var.terms.items():
            estilo = '--' if nombre in destacar else '-'
            ax.plot(var.universe, term.mf, estilo, linewidth=2, label=nombre)
        ax.set_title(titulo)
        ax.set_xlabel(var.label)
        ax.set_ylabel('Grado de pertenencia μ')
        ax.set_ylim(-0.02, 1.05)
        ax.grid(alpha=0.3)
        ax.legend(loc='center left', bbox_to_anchor=(1, 0.5))
        fig.tight_layout()
        fig.savefig(os.path.join(carpeta, archivo), dpi=150)
        plt.close(fig)

    _plot(promedio, 'Promedio (trapezoidal / triangular) + modificador "muy"',
          'mf_promedio.png', destacar=('muy_alto',))
    _plot(nivel_idioma, 'Nivel de idioma (triangulares)', 'mf_idioma.png')
    _plot(cupos, 'Cupos disponibles (gaussianas) + modificador "ligeramente"',
          'mf_cupos.png', destacar=('ligeramente_pocos',))
    _plot(idoneidad, 'Idoneidad - salida (triangulares)', 'mf_idoneidad.png')

    # Salida agregada con la línea del centroide para un caso de ejemplo
    r = evaluar(*ejemplo)
    idoneidad.view(sim=r['sim'])
    fig = plt.gcf()
    fig.set_size_inches(8, 3.8)
    plt.title(f'Salida agregada y centroide  (promedio={ejemplo[0]}, idioma={ejemplo[1]}, '
              f'cupos={ejemplo[2]}) → {r["puntaje"]}')
    fig.tight_layout()
    fig.savefig(os.path.join(carpeta, 'salida_centroide.png'), dpi=150)
    plt.close('all')


# ---------------------------------------------------------------------------
# 8. JUSTIFICACIÓN DE LA DEFUZZIFICACIÓN
# ---------------------------------------------------------------------------
# Se usa el CENTROIDE:  z* = ∫ z·μ(z) dz / ∫ μ(z) dz
#  - Considera todo el conjunto agregado (todas las reglas activadas aportan),
#    a diferencia de MOM/SOM/LOM que solo miran el máximo.
#  - Varía de forma continua y suave ante cambios pequeños en las entradas, lo que
#    evita saltos bruscos en el ranking de ofertas.
#  - Es el método estándar en Mamdani y el que mejor refleja un "compromiso"
#    entre reglas que se contradicen (p. ej. R2 dice alta y R4 dice baja).


if __name__ == '__main__':
    # [C4] La demo ya NO usa un diccionario escrito a mano: carga y razona la
    # ontología y lee de ella tanto las ofertas como los perfiles de estudiantes.
    from rdflib import Graph, Namespace, RDF
    from owlrl import DeductiveClosure, RDFS_Semantics
    g_demo = Graph()
    g_demo.parse('ontologia.ttl', format='turtle')
    DeductiveClosure(RDFS_Semantics).expand(g_demo)

    OFERTAS = ofertas_desde_grafo(g_demo)
    EXd = Namespace(NS_ONTOLOGIA)
    estudiantes = sorted(str(s).replace(NS_ONTOLOGIA, '')
                         for s in g_demo.subjects(RDF.type, EXd.Estudiante))

    for est_id in estudiantes:
        perfil = perfil_desde_grafo(g_demo, est_id)
        print(f'\n=== {est_id}  (promedio {perfil["promedio"]}, idiomas {perfil["idiomas"]}) ===')
        print(f'{"oferta":34} {"idioma":>6} {"cupos":>6} | {"puntaje":>7}  categoría')
        for oid, o in OFERTAS.items():
            r = evaluar_oferta(perfil, oid, o)
            e = r['entradas']
            cup = f'{e["cupos"]}*' if e['cupos_estimados'] else str(e['cupos'])
            print(f'{oid:34} {e["nivel_idioma"]:>6} {cup:>6} | {r["puntaje"]:>7}  {r["categoria"]}')
    print('\n* cupos no definidos en la ontología: se usa el valor neutro', CUPOS_DESCONOCIDOS)

    print('\nComparación de métodos de defuzzificación (Juan, Montpellier: promedio 4.3, francés A2=25, 4 cupos):')
    for m in ['centroid', 'bisector', 'mom']:
        print(f'  {m:9} -> {evaluar(4.3, 25, 4, metodo=m)["puntaje"]}')

    graficar_todo()
    print('\nGráficas guardadas: mf_promedio.png, mf_idioma.png, mf_cupos.png, '
          'mf_idoneidad.png, salida_centroide.png')
```

### Bloque 4 — Clases de hechos (`Fact`)

Una clase de `Fact` por cada **clase** y cada **propiedad** de la ontología, más:
- **[C8]** `Nombre` y `Descripcion` (anotaciones de los individuos),
- **[C-T1]** `EsInstanciaDe`, `Relacion` y `SubClaseDe` (respaldo genérico del traductor),
- hechos de consulta (`Hoy`, `Consulta`), el perfil derivado (`PerfilCandidato`), el resultado difuso (`ResultadoDifuso`) y los hechos de control y salida.

```python
# -*- coding: utf-8 -*-
"""
Sistema Experto (Experta) - Clases de hechos (Fact)
====================================================================
Corresponde al requisito Sec. 2.1 del enunciado (Sistema Experto) y al
Contrato Ontologia <-> Experto (Context.md, Sec. 4).

Parche de compatibilidad: `experta` depende de `frozendict`, que usa
`collections.Mapping` (retirado de la libreria estandar desde Python 3.10).
Google Colab ya corre Python >= 3.10, asi que sin este parche `import experta`
falla con AttributeError. Se aplica ANTES de importar experta.
====================================================================
"""

import collections
import collections.abc
for _name in ("Mapping", "MutableMapping", "Sequence", "Iterable", "Set"):
    if not hasattr(collections, _name):
        setattr(collections, _name, getattr(collections.abc, _name))

from datetime import date
from experta import KnowledgeEngine, Fact, Field, Rule, MATCH, NOT, TEST
from experta.strategies import DepthStrategy
from experta.agenda import Agenda


# --------------------------------------------------------------------
# 1.a Derivadas de CLASES de la ontologia -> CLASS_TO_FACT (Contrato 4.2a)
#     Un Fact simple por cada clase: solo el slot `id`.
# --------------------------------------------------------------------
class Oferta(Fact):               id = Field(str, mandatory=True)
class OfertaInternacional(Fact):  id = Field(str, mandatory=True)
class OfertaNacional(Fact):       id = Field(str, mandatory=True)
class Universidad(Fact):          id = Field(str, mandatory=True)
class UniversidadDestino(Fact):   id = Field(str, mandatory=True)
class Pais(Fact):                 id = Field(str, mandatory=True)
class AreaConocimiento(Fact):     id = Field(str, mandatory=True)
class Ingenieria(Fact):           id = Field(str, mandatory=True)
class NivelEstudios(Fact):        id = Field(str, mandatory=True)
class Convenio(Fact):             id = Field(str, mandatory=True)
class Beca(Fact):                 id = Field(str, mandatory=True)
class Estudiante(Fact):           id = Field(str, mandatory=True)

# --------------------------------------------------------------------
# 1.b Derivadas de PROPIEDADES de la ontologia -> PREDICATE_TO_FACT (4.2b)
#     Cada una lleva el sujeto y el objeto de la tripleta, tipados.
# --------------------------------------------------------------------
class TieneCupos(Fact):
    oferta = Field(str, mandatory=True); n = Field(int, mandatory=True)

class FechaCierre(Fact):
    oferta = Field(str, mandatory=True); fecha = Field(date, mandatory=True)

class OfrecidaPor(Fact):
    oferta = Field(str, mandatory=True); uni = Field(str, mandatory=True)

class OfrecidaPorUniversidadDestino(Fact):   # [C-T2] antes se descartaba esta tripleta
    oferta = Field(str, mandatory=True); uni = Field(str, mandatory=True)

class OfreceEnPais(Fact):
    oferta = Field(str, mandatory=True); pais = Field(str, mandatory=True)

class TieneArea(Fact):
    oferta = Field(str, mandatory=True); area = Field(str, mandatory=True)

class DirigidaANivel(Fact):
    oferta = Field(str, mandatory=True); nivel = Field(str, mandatory=True)

class RestringeArea(Fact):
    oferta = Field(str, mandatory=True); area = Field(str, mandatory=True)

class OfreceBeca(Fact):
    oferta = Field(str, mandatory=True); tiene = Field(bool, mandatory=True)

class RequiereIdioma(Fact):
    oferta = Field(str, mandatory=True); nivel = Field(str, mandatory=True)

class UbicadaEn(Fact):
    uni = Field(str, mandatory=True); pais = Field(str, mandatory=True)

# [C2] Hechos de las propiedades nuevas que conectan Beca, Convenio y Estudiante
class TieneBeca(Fact):
    oferta = Field(str, mandatory=True); beca = Field(str, mandatory=True)

class AmparadaPorConvenio(Fact):
    oferta = Field(str, mandatory=True); convenio = Field(str, mandatory=True)

class EstudiaArea(Fact):
    estudiante = Field(str, mandatory=True); area = Field(str, mandatory=True)

class CursaNivel(Fact):
    estudiante = Field(str, mandatory=True); nivel = Field(str, mandatory=True)

class TienePromedio(Fact):
    estudiante = Field(str, mandatory=True); valor = Field(float, mandatory=True)

class DominaIdioma(Fact):
    estudiante = Field(str, mandatory=True); idioma = Field(str, mandatory=True)

# --------------------------------------------------------------------
# 1.c [C8] Anotaciones de los individuos (nombres, títulos, descripciones)
#     Antes se descartaban; ahora se traducen y se usan en las explicaciones.
# --------------------------------------------------------------------
class Nombre(Fact):
    """foaf:name, dct:title o rdfs:label de cualquier recurso."""
    id = Field(str, mandatory=True); texto = Field(str, mandatory=True)

class Descripcion(Fact):
    """dct:description de ofertas y becas."""
    id = Field(str, mandatory=True); texto = Field(str, mandatory=True)

# --------------------------------------------------------------------
# 1.d [C-T1] Hechos GENÉRICOS de respaldo del traductor: cualquier clase o
#     propiedad ex: que NO esté en los mapas del contrato (p. ej. una propiedad
#     nueva agregada por el profesor en la prueba obligatoria) se traduce aquí
#     en lugar de perderse en silencio.
# --------------------------------------------------------------------
class EsInstanciaDe(Fact):
    id = Field(str, mandatory=True); clase = Field(str, mandatory=True)

class Relacion(Fact):
    sujeto = Field(str, mandatory=True); predicado = Field(str, mandatory=True)
    objeto = Field(object, mandatory=True)

class SubClaseDe(Fact):
    """Jerarquía de clases del esquema (rdfs:subClassOf entre clases ex:)."""
    hija = Field(str, mandatory=True); padre = Field(str, mandatory=True)

# --------------------------------------------------------------------
# 1.e Hechos NO ontologicos (no los genera el traductor)
# --------------------------------------------------------------------
class Hoy(Fact):
    """Fecha de referencia de la consulta (permite simular distintos 'hoy')."""
    fecha = Field(date, mandatory=True)

class Consulta(Fact):
    """[C9] Único dato de entrada del usuario: QUÉ estudiante (de la ontología) consulta."""
    estudiante = Field(str, mandatory=True)

class PerfilCandidato(Fact):
    """[C9] Ahora se DERIVA por la regla r00 a partir de los hechos ontológicos
    Estudiante + CursaNivel + EstudiaArea (antes se declaraba a mano)."""
    id = Field(str, mandatory=True)
    nivel_estudios = Field(str, mandatory=True)
    area = Field(str, mandatory=True)

class ResultadoDifuso(Fact):
    """Inyectado por el modulo de logica difusa (Context.md Sec. 5.2)."""
    oferta = Field(str, mandatory=True)
    puntaje = Field(float, mandatory=True)
    categoria = Field(str, mandatory=True)

class CumpleIdioma(Fact):
    """[C3] Derivado por r17: el candidato satisface el requisito de idioma de la oferta."""
    oferta = Field(str, mandatory=True); idioma = Field(str, mandatory=True)

class Descartada(Fact):
    oferta = Field(str, mandatory=True); motivo = Field(str, mandatory=True)

class Elegible(Fact):
    oferta = Field(str, mandatory=True); nivel = Field(str, mandatory=True)
    puntaje = Field(float, mandatory=True)          # [C7] para ordenar el ranking

class Procesada(Fact):
    """Marca de control (Lock-Fact) para el nivel de prioridad MEDIA."""
    oferta = Field(str, mandatory=True)

class Afinidad(Fact):
    """[C1] Afinidad parcial entre ingenierías (usa la clase Ingenieria)."""
    oferta = Field(str, mandatory=True); texto = Field(str, mandatory=True)

class Recomendacion(Fact):
    oferta = Field(str, mandatory=True)
    prioridad = Field(str, mandatory=True)
    puntaje = Field(float, mandatory=True)          # [C7]
    explicacion = Field(str, mandatory=True)

class Advertencia(Fact):
    oferta = Field(str, mandatory=True); texto = Field(str, mandatory=True)

class Respaldo(Fact):
    """[C2] La oferta está amparada por un convenio de la ontología."""
    oferta = Field(str, mandatory=True); texto = Field(str, mandatory=True)

class SugerenciaBeca(Fact):
    """[C2] Beca general (no ligada a ninguna oferta) sugerida para una oferta recomendada."""
    oferta = Field(str, mandatory=True); beca = Field(str, mandatory=True)
    texto = Field(str, mandatory=True)

class Nota(Fact):
    oferta = Field(str, mandatory=True); texto = Field(str, mandatory=True)
```

### Bloque 5 — Traductor Ontología → Sistema Experto

Recorre dinámicamente el grafo RDF **ya razonado** y genera los `Fact` iniciales. Los diccionarios mapean únicamente **vocabulario** (clases, predicados y anotaciones), nunca valores de un caso concreto.

**[C-T1]** Lo que no esté en los mapas ya no se pierde: se traduce a hechos genéricos. **[C8]** Se traducen nombres, títulos y descripciones.

```python
# -*- coding: utf-8 -*-
"""
Traductor Ontologia -> Sistema Experto
====================================================================
Recorre dinamicamente el grafo RDF YA RAZONADO (ontologia + inferencias
OWL-RL) y genera los Facts iniciales del sistema experto. Cumple la regla
del enunciado: no declara valores fijos de ningun caso concreto, solo
recorre `g` con diccionarios de VOCABULARIO (clases y predicados).

Correcciones aplicadas:
  [C-T1] Respaldo genérico: una clase o propiedad ex: que NO esté en los
         mapas ya no se pierde en silencio; se traduce a EsInstanciaDe /
         Relacion. Así la prueba obligatoria funciona también si se agrega
         una RELACIÓN NUEVA al .ttl, sin tocar Python.
  [C-T2] ex:ofrecidaPorUniversidadDestino se traduce (antes se descartaba).
  [C8]   Las anotaciones de los individuos (foaf:name, dct:title,
         rdfs:label, dct:description) se traducen a Nombre / Descripcion.
  [C-T3] valor_tipado cubre xsd:decimal y convierte de forma segura.
  El contador distingue ahora "andamiaje RDFS" de lo que sí se traduce.
====================================================================
"""

from rdflib import Namespace, Literal, URIRef
from rdflib.namespace import RDF, RDFS, XSD, FOAF, DCTERMS

EX = Namespace("http://intercambios.unal.edu.co/")

# Clase RDF -> Fact (Contrato Sec. 4.2a). Solo vocabulario, ningun valor de caso.
CLASS_TO_FACT = {
    EX.Oferta: Oferta, EX.OfertaInternacional: OfertaInternacional,
    EX.OfertaNacional: OfertaNacional, EX.Universidad: Universidad,
    EX.UniversidadDestino: UniversidadDestino, EX.Pais: Pais,
    EX.AreaConocimiento: AreaConocimiento, EX.Ingenieria: Ingenieria,
    EX.NivelEstudios: NivelEstudios, EX.Convenio: Convenio,
    EX.Beca: Beca, EX.Estudiante: Estudiante,
}

# Predicado RDF -> (Fact, nombre_slot_sujeto, nombre_slot_objeto) (Contrato 4.2b).
# Los nombres de slot son metadatos del esquema (parte del contrato), no
# valores de un caso; por eso tampoco viola la regla del traductor.
PREDICATE_TO_FACT = {
    EX.tieneCupos:     (TieneCupos, "oferta", "n"),
    EX.fechaCierre:    (FechaCierre, "oferta", "fecha"),
    EX.ofrecidaPor:    (OfrecidaPor, "oferta", "uni"),
    EX.ofrecidaPorUniversidadDestino: (OfrecidaPorUniversidadDestino, "oferta", "uni"),  # [C-T2]
    EX.ofreceEnPais:   (OfreceEnPais, "oferta", "pais"),
    EX.tieneArea:      (TieneArea, "oferta", "area"),
    EX.dirigidaANivel: (DirigidaANivel, "oferta", "nivel"),
    EX.restringeArea:  (RestringeArea, "oferta", "area"),
    EX.ofreceBeca:     (OfreceBeca, "oferta", "tiene"),
    EX.requiereIdioma: (RequiereIdioma, "oferta", "nivel"),
    EX.ubicadaEn:      (UbicadaEn, "uni", "pais"),
    # [C2] propiedades nuevas
    EX.tieneBeca:           (TieneBeca, "oferta", "beca"),
    EX.amparadaPorConvenio: (AmparadaPorConvenio, "oferta", "convenio"),
    EX.estudiaArea:         (EstudiaArea, "estudiante", "area"),
    EX.cursaNivel:          (CursaNivel, "estudiante", "nivel"),
    EX.tienePromedio:       (TienePromedio, "estudiante", "valor"),
    EX.dominaIdioma:        (DominaIdioma, "estudiante", "idioma"),
}

# [C8] Anotación -> Fact (id del recurso, texto)
ANNOTATION_TO_FACT = {
    FOAF.name: Nombre, DCTERMS.title: Nombre, RDFS.label: Nombre,
    DCTERMS.description: Descripcion,
}


def nombre_local(uri):
    """URI ex:algo -> 'algo'. Puramente mecanico, no depende de ningun caso."""
    s = str(uri)
    return s[len(str(EX)):] if s.startswith(str(EX)) else s


def es_del_dominio(uri):
    """True si el URI pertenece al namespace del proyecto (ex:)."""
    return isinstance(uri, URIRef) and str(uri).startswith(str(EX))


def valor_tipado(literal):
    """Convierte un rdflib.Literal a un tipo Python nativo segun su datatype XSD."""
    dt = literal.datatype
    try:
        if dt == XSD.integer:
            return int(literal)
        if dt in (XSD.decimal, XSD.double, XSD.float):
            return float(literal)
        if dt == XSD.date:
            return literal.toPython()
        if dt == XSD.boolean:
            return bool(literal.toPython())
    except (ValueError, TypeError):
        pass                    # literal mal formado: se conserva como texto
    return str(literal)


def traducir(g, engine):
    """
    Recorre TODAS las tripletas de `g` (ya razonado) y declara los Facts
    correspondientes en `engine`. No contiene nombres ni valores de ningun caso.

      1. rdf:type hacia una clase del contrato  -> Fact de la clase
      2. rdf:type hacia otra clase ex:           -> EsInstanciaDe  (respaldo)
      3. predicado del contrato                  -> Fact de la propiedad
      4. anotación (nombre/título/descripción)   -> Nombre / Descripcion
      5. rdfs:subClassOf entre clases ex:        -> SubClaseDe
      6. cualquier otro predicado ex:            -> Relacion       (respaldo)
      7. el resto es andamiaje de la semántica RDFS (rdfs:Resource,
         subClassOf reflexivo, declaraciones rdf:Property...) -> se ignora.
    """
    c = {"tipos": 0, "propiedades": 0, "anotaciones": 0, "esquema": 0,
         "genericos_nuevos": 0, "andamiaje_rdfs_ignorado": 0}
    for s, p, o in g:
        if p == RDF.type:
            if o in CLASS_TO_FACT:
                engine.declare(CLASS_TO_FACT[o](id=nombre_local(s))); c["tipos"] += 1
            elif es_del_dominio(o) and es_del_dominio(s):
                engine.declare(EsInstanciaDe(id=nombre_local(s), clase=nombre_local(o)))
                c["genericos_nuevos"] += 1
            else:
                c["andamiaje_rdfs_ignorado"] += 1
            continue
        if p in PREDICATE_TO_FACT:
            FactCls, slot_sujeto, slot_objeto = PREDICATE_TO_FACT[p]
            valor = valor_tipado(o) if isinstance(o, Literal) else nombre_local(o)
            try:
                engine.declare(FactCls(**{slot_sujeto: nombre_local(s), slot_objeto: valor}))
                c["propiedades"] += 1
            except Exception as err:     # tipo inesperado en el .ttl: se avisa, no se cae
                print(f"[AVISO] No se pudo traducir ({nombre_local(s)}, {nombre_local(p)}, {o}): {err}")
            continue
        if p in ANNOTATION_TO_FACT and isinstance(o, Literal) and es_del_dominio(s):
            engine.declare(ANNOTATION_TO_FACT[p](id=nombre_local(s), texto=str(o))); c["anotaciones"] += 1
            continue
        if p == RDFS.subClassOf and es_del_dominio(s) and es_del_dominio(o) and s != o:
            engine.declare(SubClaseDe(hija=nombre_local(s), padre=nombre_local(o))); c["esquema"] += 1
            continue
        if es_del_dominio(p) and es_del_dominio(s):
            valor = valor_tipado(o) if isinstance(o, Literal) else nombre_local(o)
            engine.declare(Relacion(sujeto=nombre_local(s), predicado=nombre_local(p), objeto=valor))
            c["genericos_nuevos"] += 1
            continue
        c["andamiaje_rdfs_ignorado"] += 1
    return c
```

### Bloque 6 — Motor de reglas: `SistemaExpertoIntercambios`

**Niveles de prioridad (salience):**
- **PREPARACIÓN (200):** construir el perfil del candidato desde la ontología (r00, r00b).
- **ALTA (100–110):** filtros duros → idioma (r17, r18), fecha (r01), cupos (r02), nivel (r03), área (r04, r05), consistencia (r19).
- **MEDIA (50):** elegibilidad con el resultado de la lógica difusa (r06–r09) y afinidad entre ingenierías (r20).
- **BAJA (0–10):** recomendación (r10–r12), convenio (r15, r16), becas sugeridas (r14), notas (r21), reportes (r13, r22, r23).

**[C5] Resolución de conflictos — estrategia propia `EstrategiaConflictos`.** La agenda se ordena con la clave **(salience, especificidad, recencia)**:
- **Salience:** prioridad explícita de cada `@Rule`.
- **Especificidad:** número de elementos condicionales del LHS. Ante igual salience gana la regla con más condiciones. Pares con el **mismo** salience: `r04` vs `r05`, `r07` vs `r09`, `r10`/`r11` vs `r12`.
- **Recencia:** ante empate en lo anterior, gana la activación con los hechos más recientes.

**Control de bucles (Lock-Fact):** `Descartada` (nivel ALTA), `Procesada` (nivel MEDIA) y `NOT(Recomendacion/Respaldo/...)` (nivel BAJA).

```python
# -*- coding: utf-8 -*-
"""
Motor de reglas: SistemaExpertoIntercambios
====================================================================
25 reglas (>= 15 exigidas) en 4 niveles de prioridad (salience):

  PREPARACION (200)   -> construir el perfil del candidato desde la ontología.
  ALTA  (100-110)     -> filtros duros: descartar ofertas que no aplican.
  MEDIA (50)          -> elegibilidad: usa el resultado de la lógica difusa.
  BAJA  (0-10)        -> salida: recomendación, respaldo, avisos y reportes.

Mecanismos de resolución de conflictos (los 3 exigidos):

  [C5] Se implementa una ESTRATEGIA PROPIA (EstrategiaConflictos) que ordena
  la agenda con la clave  (salience, especificidad, recencia):

  - SALIENCE: prioridad explícita de cada @Rule. Decide qué NIVEL corre primero.
  - ESPECIFICIDAD: número de elementos condicionales (patrones, NOT, TEST) de
    la regla. Ante igual salience gana la regla con MÁS condiciones. Antes se
    simulaba subiendo el salience; ahora la calcula la estrategia, así que los
    pares general/específica tienen el MISMO salience:
        r04 (general) vs r05 (exclusiva)          -> ambas salience 100
        r07 (media)   vs r09 (afinidad exacta)    -> ambas salience 50
        r10/r11       vs r12 (con beca)           -> ambas salience 10
  - RECENCIA: ante igual salience y especificidad, gana la activación cuyos
    hechos son MÁS RECIENTES (mayor __factid__), igual que DepthStrategy.

  La agenda (AgendaConTraza) registra cada disparo con su clave, para poder
  EVIDENCIAR en la ejecución qué mecanismo decidió (ver celda de evidencia).

Control de bucles / duplicados (Lock-Fact):
  (a) ALTA : el hecho `Descartada` es el candado (todas exigen NOT(Descartada)).
  (b) MEDIA: hecho de control `Procesada(oferta)`.
  (c) BAJA : NOT(Recomendacion(...)), NOT(SugerenciaBeca(...)), etc.

[C13] Las condiciones que antes estaban dentro del cuerpo de la regla
(`if f < hoy`, `if ac != a`) se movieron al LHS con TEST(...): ahora la regla
solo se activa cuando la condición se cumple.
====================================================================
"""


# ------------------------------------------------------------------
# [C5] Estrategia de resolución de conflictos: salience > especificidad > recencia
# ------------------------------------------------------------------
class EstrategiaConflictos(DepthStrategy):
    """Ordena la agenda por (salience, especificidad, recencia).
    La activación con la clave MAYOR es la siguiente en dispararse."""

    def get_key(self, activation):
        salience = activation.rule.salience
        especificidad = len(activation.rule)      # nº de elementos condicionales del LHS
        recencia = sorted((f['__factid__'] for f in activation.facts), reverse=True)
        return (salience, especificidad, recencia)


class AgendaConTraza(Agenda):
    """Agenda que guarda la secuencia de disparos (regla + clave de resolución)."""

    def __init__(self):
        super().__init__()
        self.traza = []

    def get_next(self):
        act = super().get_next()
        if act is not None:
            sal, esp, rec = act.key
            self.traza.append((act.rule.__name__, sal, esp, rec[:3]))
        return act


SIN_CONVENIO = "sin convenio registrado en la ontología: confirmar convenio vigente con la DRE"


class SistemaExpertoIntercambios(KnowledgeEngine):

    __strategy__ = EstrategiaConflictos

    def __init__(self):
        super().__init__()
        self.agenda = AgendaConTraza()

    def reset(self, **kwargs):
        # reset() de Experta crea una Agenda nueva: se reemplaza por la que guarda la traza.
        super().reset(**kwargs)
        self.agenda = AgendaConTraza()

    # ============== PREPARACIÓN (salience 200): perfil desde la ontología ==============
    @Rule(
        Consulta(estudiante=MATCH.e),
        Estudiante(id=MATCH.e),
        CursaNivel(estudiante=MATCH.e, nivel=MATCH.n),
        EstudiaArea(estudiante=MATCH.e, area=MATCH.a),
        NOT(PerfilCandidato(id=MATCH.e)),
        salience=200,
    )
    def r00_perfil_desde_ontologia(self, e, n, a):
        """[C9] El perfil del candidato se deriva de la ontología (no se escribe a mano)."""
        self.declare(PerfilCandidato(id=e, nivel_estudios=n, area=a))

    @Rule(
        Consulta(estudiante=MATCH.e),
        NOT(Estudiante(id=MATCH.e)),
        salience=200,
    )
    def r00b_estudiante_inexistente(self, e):
        print(f"   [error] el estudiante '{e}' no existe en la ontología: no se puede evaluar")

    # ================= ALTA (salience 100-110): filtros duros =================
    @Rule(
        Consulta(estudiante=MATCH.e),
        RequiereIdioma(oferta=MATCH.o, nivel=MATCH.req),
        DominaIdioma(estudiante=MATCH.e, idioma=MATCH.dom),
        TEST(lambda req, dom: cumple_idioma(req, dom)),
        NOT(CumpleIdioma(oferta=MATCH.o)),
        salience=110,
    )
    def r17_cumple_idioma(self, o, dom):
        """[C3] Deriva que el candidato cumple el requisito de idioma de la oferta."""
        self.declare(CumpleIdioma(oferta=o, idioma=dom))

    @Rule(
        Hoy(fecha=MATCH.hoy),
        Oferta(id=MATCH.o),
        FechaCierre(oferta=MATCH.o, fecha=MATCH.f),
        TEST(lambda f, hoy: f < hoy),                       # [C13] condición en el LHS
        NOT(Descartada(oferta=MATCH.o)),
        salience=100,
    )
    def r01_fecha_vencida(self, o, f):
        self.declare(Descartada(oferta=o, motivo=f"convocatoria cerrada (cierre {f})"))

    @Rule(
        Oferta(id=MATCH.o),
        TieneCupos(oferta=MATCH.o, n=0),
        NOT(Descartada(oferta=MATCH.o)),
        salience=100,
    )
    def r02_sin_cupos(self, o):
        self.declare(Descartada(oferta=o, motivo="sin cupos disponibles"))

    @Rule(
        Oferta(id=MATCH.o),
        PerfilCandidato(nivel_estudios=MATCH.n),
        NivelEstudios(id=MATCH.n),
        NOT(DirigidaANivel(oferta=MATCH.o, nivel=MATCH.n)),
        NOT(Descartada(oferta=MATCH.o)),
        salience=100,
    )
    def r03_nivel_no_ofrecido(self, o, n):
        self.declare(Descartada(oferta=o, motivo=f"no dirigida al nivel '{n}' del candidato"))

    @Rule(
        Oferta(id=MATCH.o),
        RestringeArea(oferta=MATCH.o, area=MATCH.a),
        PerfilCandidato(area=MATCH.ac),
        TEST(lambda ac, a: ac != a),                        # [C13]
        NOT(Descartada(oferta=MATCH.o)),
        salience=100,
    )
    def r04_area_restringida_general(self, o, a):
        self.declare(Descartada(oferta=o, motivo=f"restringida al area '{a}'"))

    @Rule(
        Oferta(id=MATCH.o),
        RestringeArea(oferta=MATCH.o, area=MATCH.a),
        TieneArea(oferta=MATCH.o, area=MATCH.a),          # condición extra -> más específica
        PerfilCandidato(area=MATCH.ac),
        TEST(lambda ac, a: ac != a),
        NOT(Descartada(oferta=MATCH.o)),
        salience=100,                                      # [C5] MISMO salience que r04
    )
    def r05_area_restringida_exclusiva(self, o, a):
        self.declare(Descartada(
            oferta=o,
            motivo=f"oferta EXCLUSIVA del area '{a}' (no solo restringida, es su unica area)"))

    @Rule(
        Consulta(estudiante=MATCH.e),
        Oferta(id=MATCH.o),
        RequiereIdioma(oferta=MATCH.o, nivel=MATCH.req),
        NOT(CumpleIdioma(oferta=MATCH.o)),
        NOT(Descartada(oferta=MATCH.o)),
        salience=100,
    )
    def r18_idioma_insuficiente(self, o, req):
        """[C3] Filtro duro: el candidato no alcanza el idioma exigido (usa RequiereIdioma)."""
        self.declare(Descartada(oferta=o, motivo=f"no cumple el requisito de idioma ({req})"))

    @Rule(
        OfrecidaPor(oferta=MATCH.o, uni=MATCH.u),
        UbicadaEn(uni=MATCH.u, pais=MATCH.p1),
        OfreceEnPais(oferta=MATCH.o, pais=MATCH.p2),
        TEST(lambda p1, p2: p1 != p2),
        NOT(Advertencia(oferta=MATCH.o, texto="datos inconsistentes")),
        salience=100,
    )
    def r19_ubicacion_inconsistente(self, o, u, p1, p2):
        """[C1] Control de calidad con UbicadaEn: la universidad y la oferta deben estar en el mismo país."""
        self.declare(Advertencia(oferta=o, texto="datos inconsistentes"))
        print(f"   [aviso] {o}: la universidad {u} está en {p1} pero la oferta dice {p2}")

    # ============= MEDIA (salience 50): elegibilidad (usa difuso) ==========
    @Rule(
        ResultadoDifuso(oferta=MATCH.o, categoria="alta", puntaje=MATCH.p),
        NOT(Descartada(oferta=MATCH.o)),
        NOT(Procesada(oferta=MATCH.o)),
        salience=50,
    )
    def r06_elegible_alta(self, o, p):
        self.declare(Elegible(oferta=o, nivel="alta", puntaje=p))
        self.declare(Procesada(oferta=o))

    @Rule(
        ResultadoDifuso(oferta=MATCH.o, categoria="media", puntaje=MATCH.p),
        NOT(Descartada(oferta=MATCH.o)),
        NOT(Procesada(oferta=MATCH.o)),
        salience=50,
    )
    def r07_elegible_media(self, o, p):
        self.declare(Elegible(oferta=o, nivel="media", puntaje=p))
        self.declare(Procesada(oferta=o))

    @Rule(
        ResultadoDifuso(oferta=MATCH.o, categoria="baja"),
        NOT(Descartada(oferta=MATCH.o)),
        NOT(Procesada(oferta=MATCH.o)),
        salience=50,
    )
    def r08_idoneidad_baja_descarta(self, o):
        self.declare(Descartada(oferta=o, motivo="idoneidad difusa baja (perfil/idioma insuficiente)"))
        self.declare(Procesada(oferta=o))

    @Rule(
        ResultadoDifuso(oferta=MATCH.o, categoria="media", puntaje=MATCH.p),
        TieneArea(oferta=MATCH.o, area=MATCH.a),
        PerfilCandidato(area=MATCH.a),                     # el candidato coincide EXACTO con el área
        AreaConocimiento(id=MATCH.a),
        NOT(Descartada(oferta=MATCH.o)),
        NOT(Procesada(oferta=MATCH.o)),
        salience=50,                                        # [C5] MISMO salience que r07
    )
    def r09_afinidad_area_exacta(self, o, a, p):
        self.declare(Elegible(oferta=o, nivel="alta", puntaje=p))
        self.declare(Procesada(oferta=o))
        self.declare(Nota(oferta=o, texto=f"sube a prioridad alta por afinidad exacta de área ({a})"))

    @Rule(
        TieneArea(oferta=MATCH.o, area=MATCH.a2),
        Ingenieria(id=MATCH.a2),
        PerfilCandidato(area=MATCH.ac),
        Ingenieria(id=MATCH.ac),
        TEST(lambda ac, a2: ac != a2),
        NOT(Descartada(oferta=MATCH.o)),
        NOT(Afinidad(oferta=MATCH.o)),
        salience=50,
    )
    def r20_afinidad_ingenieria(self, o, a2, ac):
        """[C1] Usa la subclase Ingenieria: afinidad parcial entre ingenierías distintas."""
        self.declare(Afinidad(oferta=o, texto=f"afinidad parcial: ambas son ingenierías ({ac} / {a2})"))

    # =================== BAJA (salience 0-10): salida =========================
    @Rule(
        Elegible(oferta=MATCH.o, nivel=MATCH.nv, puntaje=MATCH.p),
        OfertaInternacional(id=MATCH.o),
        OfrecidaPor(oferta=MATCH.o, uni=MATCH.u),
        Universidad(id=MATCH.u),
        UbicadaEn(uni=MATCH.u, pais=MATCH.pa),             # [C1] usa UbicadaEn
        Pais(id=MATCH.pa),
        Nombre(id=MATCH.u, texto=MATCH.nu),                 # [C8] nombres legibles
        Nombre(id=MATCH.pa, texto=MATCH.npa),
        NOT(Recomendacion(oferta=MATCH.o)),
        salience=10,
    )
    def r10_recomendacion_internacional(self, o, nv, p, nu, npa):
        self.declare(Recomendacion(
            oferta=o, prioridad=nv, puntaje=p,
            explicacion=f"Intercambio internacional en {npa} con {nu}; idoneidad {nv} ({p})."))

    @Rule(
        Elegible(oferta=MATCH.o, nivel=MATCH.nv, puntaje=MATCH.p),
        OfertaNacional(id=MATCH.o),
        OfrecidaPorUniversidadDestino(oferta=MATCH.o, uni=MATCH.u),
        UniversidadDestino(id=MATCH.u),
        Nombre(id=MATCH.u, texto=MATCH.nu),
        NOT(Recomendacion(oferta=MATCH.o)),
        salience=10,
    )
    def r11_recomendacion_nacional(self, o, nv, p, nu):
        self.declare(Recomendacion(
            oferta=o, prioridad=nv, puntaje=p,
            explicacion=f"Intercambio nacional con {nu}; idoneidad {nv} ({p})."))

    @Rule(
        Elegible(oferta=MATCH.o, nivel="alta", puntaje=MATCH.p),
        OfreceBeca(oferta=MATCH.o, tiene=True),
        TieneBeca(oferta=MATCH.o, beca=MATCH.b),            # [C2] usa Beca de verdad
        Beca(id=MATCH.b),
        Nombre(id=MATCH.b, texto=MATCH.nb),
        OfrecidaPor(oferta=MATCH.o, uni=MATCH.u),
        Universidad(id=MATCH.u),
        UbicadaEn(uni=MATCH.u, pais=MATCH.pa),
        Pais(id=MATCH.pa),
        Nombre(id=MATCH.u, texto=MATCH.nu),
        NOT(Recomendacion(oferta=MATCH.o)),
        salience=10,                                        # [C5] MISMO salience que r10/r11
    )
    def r12_recomendacion_con_beca(self, o, p, nb, nu):
        self.declare(Recomendacion(
            oferta=o, prioridad="alta+beca", puntaje=p,
            explicacion=f"Idoneidad alta con {nu} y BECA '{nb}': prioridad máxima ({p})."))

    @Rule(
        Recomendacion(oferta=MATCH.o),
        AmparadaPorConvenio(oferta=MATCH.o, convenio=MATCH.c),   # [C2] usa Convenio
        Convenio(id=MATCH.c),
        Nombre(id=MATCH.c, texto=MATCH.nc),
        NOT(Respaldo(oferta=MATCH.o)),
        salience=5,
    )
    def r15_respaldo_convenio(self, o, nc):
        self.declare(Respaldo(oferta=o, texto=f"respaldada por: {nc}"))

    @Rule(
        Recomendacion(oferta=MATCH.o),
        NOT(AmparadaPorConvenio(oferta=MATCH.o)),
        NOT(Advertencia(oferta=MATCH.o, texto=SIN_CONVENIO)),
        salience=5,
    )
    def r16_advertencia_sin_convenio(self, o):
        self.declare(Advertencia(oferta=o, texto=SIN_CONVENIO))

    @Rule(
        Recomendacion(oferta=MATCH.o),
        OfertaInternacional(id=MATCH.o),
        NOT(TieneBeca(oferta=MATCH.o)),
        Beca(id=MATCH.b),
        NOT(TieneBeca(beca=MATCH.b)),                      # beca general: no ligada a ninguna oferta
        Nombre(id=MATCH.b, texto=MATCH.nb),
        NOT(SugerenciaBeca(oferta=MATCH.o, beca=MATCH.b)),
        salience=5,
    )
    def r14_sugerir_beca_general(self, o, b, nb):
        """[C2] Beca de la ontología sugerida para ofertas internacionales sin beca propia."""
        self.declare(SugerenciaBeca(oferta=o, beca=b, texto=nb))

    @Rule(
        Recomendacion(oferta=MATCH.o),
        Descripcion(id=MATCH.o, texto=MATCH.d),              # [C8] usa dct:description
        NOT(Nota(oferta=MATCH.o, texto=MATCH.d)),
        salience=5,
    )
    def r21_nota_descripcion(self, o, d):
        self.declare(Nota(oferta=o, texto=d))

    @Rule(Descartada(oferta=MATCH.o, motivo=MATCH.m), Nombre(id=MATCH.o, texto=MATCH.t), salience=0)
    def r13_informe_descarte(self, o, m, t):
        print(f"   [descartada] {t}: {m}")

    # ------- [C-T1] Reporte de conocimiento NUEVO que no está en el contrato -------
    @Rule(EsInstanciaDe(id=MATCH.i, clase=MATCH.c), SubClaseDe(hija=MATCH.c, padre=MATCH.sup), salience=0)
    def r22_clase_nueva(self, i, c, sup):
        print(f"   [nuevo] '{i}' es instancia de la clase nueva '{c}' (subclase de '{sup}')")

    @Rule(Relacion(sujeto=MATCH.s, predicado=MATCH.p, objeto=MATCH.v), salience=0)
    def r23_relacion_nueva(self, s, p, v):
        print(f"   [nuevo] relación no prevista en el contrato: {s} --{p}--> {v}")
```

### Bloque 7 — Integración completa: Ontología (razonada) + Lógica difusa (real) → Sistema Experto

**[C9]** La única entrada es **qué estudiante** de la ontología consulta y la fecha. Todo lo demás (perfil, ofertas, cupos, idiomas) sale del grafo razonado.

| Escenario | Estudiante | Qué demuestra |
|---|---|---|
| A | Juan (Ing. Sistemas) | exclusión por área (r05), por idioma (r18), afinidad exacta (r09) y beca (r12) |
| B | Juan, otra fecha | convocatorias cerradas (r01) |
| C | Carlos (maestría) | nivel no ofrecido (r03) |
| D | María (Matemáticas) | la **misma** oferta (Stavanger) que se descartaba ahora se recomienda |
| E | Ana (Humanidades) | sin el idioma exigido (r18) y promedio bajo |

```python
# -*- coding: utf-8 -*-
"""
Integracion completa: Ontologia (razonada) + Logica difusa (real) -> Sistema Experto
====================================================================
Pipeline exacto exigido por el enunciado (Context.md Sec. 5.1.1):
  1. g.parse(ontologia.ttl)                      # cargar la fuente canonica
  2. DeductiveClosure(RDFS_Semantics).expand(g)   # razonar
  3. traducir(g, engine)                         # recorrer g YA razonado -> Facts
  4. perfil_desde_grafo + ofertas_desde_grafo    # [C9] TODO sale del grafo
     + resultados_para_experto(...)              # logica difusa REAL
  5. declarar Consulta(estudiante) + ResultadoDifuso
  6. engine.run()                                # disparar las reglas

[C9] La única entrada del usuario es QUÉ estudiante consulta (su id en la
ontología) y la fecha. El perfil (área, nivel, promedio, idiomas) sale del grafo.
[C7] El reporte ordena las recomendaciones por PUNTAJE difuso (antes se
ordenaba por el texto de la prioridad y "media" quedaba antes que "alta").
====================================================================
"""

g = Graph()
g.parse("ontologia.ttl", format="turtle")
print(f"Triples antes de razonar : {len(g)}")
DeductiveClosure(RDFS_Semantics).expand(g)
print(f"Triples despues de razonar: {len(g)}")

ORDEN_PRIORIDAD = {"alta+beca": 3, "alta": 2, "media": 1}


def hechos_de(engine, clase):
    """Lista los hechos de una clase presentes en la memoria del motor."""
    return [f for f in engine.facts.values() if isinstance(f, clase)]


def reporte(engine):
    """Imprime las recomendaciones ordenadas (prioridad y puntaje) con sus notas."""
    extras = {}
    for clase, etiqueta in ((Respaldo, "respaldo"), (Advertencia, "aviso"),
                            (SugerenciaBeca, "beca sugerida"), (Nota, "nota"), (Afinidad, "afinidad")):
        for f in hechos_de(engine, clase):
            extras.setdefault(f["oferta"], []).append(f"{etiqueta}: {f['texto']}")
    recs = hechos_de(engine, Recomendacion)
    recs.sort(key=lambda f: (ORDEN_PRIORIDAD.get(f["prioridad"], 0), f["puntaje"]), reverse=True)
    print("\n--- RECOMENDACIONES (ordenadas por prioridad y puntaje difuso) ---")
    if not recs:
        print("  (ninguna oferta recomendada)")
    for i, r in enumerate(recs, 1):
        print(f"  {i}. [{r['prioridad']:9}] {r['explicacion']}")
        for e in extras.get(r["oferta"], []):
            print(f"        · {e}")


def correr_caso(g, hoy, estudiante_id, titulo, verbose=True):
    """Ejecuta el pipeline completo para UN estudiante de la ontología."""
    print(f"\n{'='*76}\n  {titulo}\n  hoy={hoy}  estudiante={estudiante_id}\n{'='*76}")

    # ---- Lógica difusa REAL: perfil y ofertas leídos del grafo razonado ----
    ofertas = ofertas_desde_grafo(g)
    resultados = []
    try:
        perfil = perfil_desde_grafo(g, estudiante_id)
        print(f"Perfil leído de la ontología: {perfil}")
        resultados = resultados_para_experto(perfil, ofertas)
    except ValueError as err:
        print(f"[AVISO] {err}")

    # ---- Traductor + motor de reglas ----
    engine = SistemaExpertoIntercambios()
    engine.reset()
    stats = traducir(g, engine)
    if verbose:
        print("Traductor ->", stats)

    engine.declare(Hoy(fecha=hoy))
    engine.declare(Consulta(estudiante=estudiante_id))
    for r in resultados:
        engine.declare(ResultadoDifuso(**r))

    print(f"Hechos declarados antes de correr: {len(engine.facts)}  (minimo exigido: 40)")
    engine.run()
    reporte(engine)
    print(f"\nTotal hechos finales: {len(engine.facts)}  |  reglas disparadas: {len(engine.agenda.traza)}")
    return engine


# Escenario A: Juan (Ing. Sistemas, pregrado, 4.3, inglés B2, francés A2), nada vencido.
#  -> Stavanger: exclusiva de Matemáticas (r05)   -> Montpellier: francés A2 < B1 (r18)
#  -> Escuela Colombiana: afinidad exacta de área (r09) + beca ICETEX (r12)
eng_A = correr_caso(g, date(2026, 8, 1), "est_juan_sistemas",
                    "ESCENARIO A: Juan — área, idioma, afinidad y beca")

# Escenario B: mismo estudiante más tarde -> fecha vencida (r01).
eng_B = correr_caso(g, date(2026, 9, 20), "est_juan_sistemas",
                    "ESCENARIO B: Juan en otra fecha — convocatorias cerradas (r01)")

# Escenario C: Carlos (Arquitectura, MAESTRÍA) -> nivel no ofrecido (r03).
eng_C = correr_caso(g, date(2026, 8, 1), "est_carlos_arquitectura",
                    "ESCENARIO C: Carlos (maestría) — nivel no ofrecido (r03)")

# Escenario D: María (Matemáticas, 4.7, inglés C1) -> Stavanger YA NO se descarta
#  (es su área) y Freie Berlin sale con beca FUB (r12 internacional).
eng_D = correr_caso(g, date(2026, 8, 1), "est_maria_matematicas",
                    "ESCENARIO D: María — la MISMA oferta (Stavanger) ahora se recomienda")

# Escenario E: Ana (Humanidades, 3.1, solo portugués B1) -> las internacionales que exigen
#  inglés/francés se descartan por idioma (r18) y el resto queda con idoneidad media.
eng_E = correr_caso(g, date(2026, 8, 1), "est_ana_humanidades",
                    "ESCENARIO E: Ana — sin el idioma exigido (r18) y promedio bajo")
```

### Pruebas unitarias de reglas sin cobertura en los datos reales

Algunas reglas no se activan con los 8 datos reales (0 cupos, área parcialmente restringida, idoneidad baja sin otro descarte, datos inconsistentes, otra ingeniería, estudiante inexistente). Se prueban con hechos sintéticos, que es lo normal en una prueba unitaria.

```python
# -*- coding: utf-8 -*-
"""
Pruebas unitarias: reglas que NO se disparan con los datos reales
====================================================================
Ninguna oferta real tiene 0 cupos, un área PARCIALMENTE restringida,
datos de ubicación inconsistentes, idoneidad baja sin otro motivo de
descarte, ni un candidato de otra ingeniería.
Se verifica cada una de esas reglas con hechos sintéticos (lo normal en
una prueba unitaria) para no dejarla sin evidencia de que funciona.
[C6] Se agregan pruebas para r08, r19, r20 y r00b.
====================================================================
"""

print(f"{'='*72}\n  PRUEBAS UNITARIAS DE REGLAS SIN COBERTURA EN LOS DATOS REALES\n{'='*72}")


def motor_con(*hechos):
    """Crea un motor limpio, declara los hechos sintéticos y lo ejecuta."""
    eng = SistemaExpertoIntercambios()
    eng.reset()
    for h in hechos:
        eng.declare(h)
    eng.run()
    return eng


# r02: oferta con 0 cupos -> debe descartarse.
eng = motor_con(Oferta(id="oferta_prueba_cupos"), TieneCupos(oferta="oferta_prueba_cupos", n=0))
ok02 = any(f["oferta"] == "oferta_prueba_cupos" for f in hechos_de(eng, Descartada))
print("r02_sin_cupos                  disparó correctamente:", ok02)
assert ok02

# r04: oferta con área 'area_x' restringida a 'area_y' (no exclusiva) -> regla GENERAL.
eng = motor_con(Oferta(id="oferta_prueba_area"),
                TieneArea(oferta="oferta_prueba_area", area="area_x"),
                RestringeArea(oferta="oferta_prueba_area", area="area_y"),
                PerfilCandidato(id="c", nivel_estudios="nivel_pregrado", area="area_z"))
descartes = hechos_de(eng, Descartada)
ok04 = len(descartes) == 1 and descartes[0]["motivo"].startswith("restringida al area")
print("r04_area_restringida_general   disparó correctamente:", ok04, "->", descartes[0]["motivo"])
assert ok04

# r08: categoría difusa 'baja' -> la oferta se descarta por idoneidad.
#  (Con los datos reales no ocurre: las ofertas con idoneidad baja ya fueron
#   descartadas antes por idioma o área, que tienen mayor prioridad.)
eng = motor_con(ResultadoDifuso(oferta="oferta_prueba_baja", puntaje=20.0, categoria="baja"))
ok08 = any(f["oferta"] == "oferta_prueba_baja" and "idoneidad" in f["motivo"] for f in hechos_de(eng, Descartada))
print("r08_idoneidad_baja_descarta    disparó correctamente:", ok08)
assert ok08

# r19: la universidad está en un país distinto al de la oferta -> advertencia.
eng = motor_con(OfrecidaPor(oferta="oferta_x", uni="uni_x"),
                UbicadaEn(uni="uni_x", pais="pais_a"),
                OfreceEnPais(oferta="oferta_x", pais="pais_b"))
ok19 = any(f["texto"] == "datos inconsistentes" for f in hechos_de(eng, Advertencia))
print("r19_ubicacion_inconsistente    disparó correctamente:", ok19)
assert ok19

# r20: candidato de Ing. Civil frente a oferta de Ing. de Sistemas -> afinidad parcial.
eng = motor_con(TieneArea(oferta="oferta_y", area="area_ingenieria_sistemas"),
                Ingenieria(id="area_ingenieria_sistemas"), Ingenieria(id="area_ingenieria_civil"),
                PerfilCandidato(id="c", nivel_estudios="nivel_pregrado", area="area_ingenieria_civil"))
afin = hechos_de(eng, Afinidad)
ok20 = len(afin) == 1
print("r20_afinidad_ingenieria        disparó correctamente:", ok20, "->", afin[0]["texto"] if afin else None)
assert ok20

# r00b: consulta por un estudiante que no existe en la ontología -> error controlado.
eng = motor_con(Consulta(estudiante="est_no_existe"))
ok00b = not hechos_de(eng, PerfilCandidato)
print("r00b_estudiante_inexistente    disparó correctamente:", ok00b)
assert ok00b

print("\n[OK] Las 6 pruebas unitarias pasaron.")
```

### [C5] Evidencia de los mecanismos de resolución de conflictos

La agenda registra cada disparo con su clave `(salience, especificidad, recencia)`. Esta celda muestra, con casos mínimos y con la traza real del escenario A, qué mecanismo decidió en cada caso.

```python
# -*- coding: utf-8 -*-
"""
[C5] EVIDENCIA de los 3 mecanismos de resolución de conflictos
====================================================================
La agenda registra cada disparo con su clave (salience, especificidad,
recencia). Aquí se muestra, con casos mínimos, qué mecanismo decidió.
====================================================================
"""


def mostrar_traza(eng, filtro=None):
    """Imprime la secuencia de disparos: regla, salience, especificidad y hechos más recientes."""
    for i, (regla, sal, esp, rec) in enumerate(eng.agenda.traza, 1):
        if filtro is None or any(regla.startswith(f) for f in filtro):
            print(f"   {i:>2}. {regla:32} salience={sal:<4} especificidad={esp:<3} hechos={rec}")


# ---------------- 1. SALIENCE ----------------
print("1) SALIENCE — para Stavanger están activadas a la vez r05 (salience 100) y r07 (salience 50).")
print("   Dispara primero r05 (mayor salience); r07 queda bloqueada por NOT(Descartada):")
eng = motor_con(Oferta(id="of_s"), TieneArea(oferta="of_s", area="area_mat"),
                RestringeArea(oferta="of_s", area="area_mat"),
                PerfilCandidato(id="c", nivel_estudios="nivel_pregrado", area="area_sis"),
                ResultadoDifuso(oferta="of_s", puntaje=55.0, categoria="media"))
mostrar_traza(eng, ("r05", "r07", "r04"))
assert not hechos_de(eng, Elegible)

# ---------------- 2. ESPECIFICIDAD ----------------
print("\n2) ESPECIFICIDAD — r04 (5 condiciones) y r05 (6 condiciones) tienen el MISMO salience (100).")
print("   Gana r05 por ser más específica; r04 nunca dispara (su activación se invalida):")
mostrar_traza(eng, ("r04", "r05"))
assert [t[0] for t in eng.agenda.traza if t[0].startswith(("r04", "r05"))] == ["r05_area_restringida_exclusiva"]

# ---------------- 3. RECENCIA ----------------
print("\n3) RECENCIA — dos ofertas vencidas activan r01 con el mismo salience y especificidad.")
print("   Dispara primero la que usa el hecho declarado MÁS RECIENTEMENTE (of_nueva):")
eng = motor_con(Hoy(fecha=date(2026, 9, 20)),
                Oferta(id="of_vieja"), FechaCierre(oferta="of_vieja", fecha=date(2026, 9, 1)),
                Oferta(id="of_nueva"), FechaCierre(oferta="of_nueva", fecha=date(2026, 9, 2)))
mostrar_traza(eng, ("r01",))
orden = [f["oferta"] for f in sorted(hechos_de(eng, Descartada), key=lambda f: f["__factid__"])]
print("   Orden en que quedaron descartadas:", orden)
assert orden == ["of_nueva", "of_vieja"]

# ---------------- 4. Traza real del escenario A ----------------
print("\n4) Primeros 12 disparos del ESCENARIO A (sistema completo):")
for i, (regla, sal, esp, rec) in enumerate(eng_A.agenda.traza[:12], 1):
    print(f"   {i:>2}. {regla:32} salience={sal:<4} especificidad={esp:<3} hechos={rec}")

print("\n[OK] Los tres mecanismos quedan evidenciados en la ejecución.")
```

### Prueba obligatoria del traductor (enunciado: "Caso de prueba obligatorio")

Se agrega **únicamente en `ontologia.ttl`** una oferta nueva **y una relación nueva** (`ex:ofreceAlojamiento`), sin tocar ninguna línea de Python. Con el mismo código se evidencia que la oferta aparece como `Fact`, la califica la lógica difusa y la procesan las reglas, y que la relación nueva se traduce como hecho genérico.

**[C15]** Al terminar, el `.ttl` se restaura automáticamente a su versión canónica.

```python
# -*- coding: utf-8 -*-
"""
PRUEBA OBLIGATORIA DEL TRADUCTOR (enunciado: "Caso de prueba obligatorio")
====================================================================
Se agrega al archivo `ontologia.ttl` (sin tocar ni una línea de Python):
  (a) un INDIVIDUO nuevo: una oferta y su universidad, y
  (b) una RELACIÓN NUEVA que no existe en el contrato: ex:ofreceAlojamiento.
Se vuelve a razonar y traducir con el MISMO código y se evidencia que:
  - la oferta nueva aparece como Fact y la procesa la lógica difusa y las reglas;
  - la relación nueva no se pierde: el traductor la convierte en `Relacion`
    y la regla r23 la reporta.

[C4] El resultado difuso de la oferta nueva ya NO se escribe a mano: se
calcula con ofertas_desde_grafo(g2) + perfil_desde_grafo(g2, ...).
[C15] Al final se RESTAURA automáticamente el .ttl canónico, para que el
archivo que va en el ZIP no quede con la oferta de prueba.
====================================================================
"""

NUEVO_INDIVIDUO_TTL = """
ex:oferta_prueba_2027_1 a ex:OfertaNacional ;
    ex:dirigidaANivel ex:nivel_pregrado ;
    ex:fechaCierre "2026-12-01"^^xsd:date ;
    ex:ofreceEnPais ex:pais_colombia ;
    ex:ofrecidaPorUniversidadDestino ex:uni_prueba_2027_1 ;
    ex:tieneArea ex:area_comun ;
    ex:tieneCupos 3 ;
    ex:ofreceAlojamiento true ;
    dct:title "Oferta de prueba agregada solo en el ttl (sin tocar Python)"@es .

ex:uni_prueba_2027_1 a ex:UniversidadDestino ;
    ex:ubicadaEn ex:pais_colombia ;
    foaf:name "Universidad de Prueba" .
"""

with open("ontologia.ttl", encoding="utf-8") as f:
    TTL_CANONICO = f.read()                      # copia para restaurar al final

try:
    with open("ontologia.ttl", "a", encoding="utf-8") as f:
        f.write(NUEVO_INDIVIDUO_TTL)
    print("[OK] Individuo y relación nuevos anexados a ontologia.ttl (edición NO en Python).")

    # Mismo pipeline, mismo traductor, mismo motor: ningún cambio de código.
    g2 = Graph()
    g2.parse("ontologia.ttl", format="turtle")
    DeductiveClosure(RDFS_Semantics).expand(g2)

    engine2 = correr_caso(g2, date(2026, 8, 1), "est_juan_sistemas",
                          "PRUEBA OBLIGATORIA: oferta y relación nuevas solo en el .ttl")

    presente = any(f["id"] == "oferta_prueba_2027_1" for f in hechos_de(engine2, OfertaNacional))
    difuso = [f for f in hechos_de(engine2, ResultadoDifuso) if f["oferta"] == "oferta_prueba_2027_1"]
    relacion = [f for f in hechos_de(engine2, Relacion) if f["predicado"] == "ofreceAlojamiento"]
    rec_nueva = [f for f in hechos_de(engine2, Recomendacion) if f["oferta"] == "oferta_prueba_2027_1"]

    print("\nVerificación:")
    print("  Fact OfertaNacional(oferta_prueba_2027_1) generado automáticamente:", presente)
    print("  ResultadoDifuso calculado desde el grafo:", dict(difuso[0]) if difuso else None)
    print("  Relación nueva traducida como Fact genérico:", dict(relacion[0]) if relacion else None)
    print("  Recomendación generada por las reglas:", dict(rec_nueva[0]) if rec_nueva else None)
    assert presente and difuso and relacion and rec_nueva
    print("\n[OK] La prueba obligatoria se cumple: el cambio quedó SOLO en ontologia.ttl y se"
          " reflejó automáticamente como Facts, en la lógica difusa y en las reglas.")
finally:
    with open("ontologia.ttl", "w", encoding="utf-8") as f:
        f.write(TTL_CANONICO)
    print("[OK] ontologia.ttl restaurado a su versión canónica.")
```
