# Workflow con Gemini: Exploración de Oportunidades de IA

En este cuaderno usaremos la familia de modelos **Gemini 3** de Google para crear
un workflow de múltiples pasos donde el modelo genera ideas encadenadas.

### ¿Qué es Gemini?
- Gemini es la familia de modelos de IA generativa de **Google**.
- Cada generación incluye modelos de distinto tamaño. Por ejemplo, `gemini-3.1-flash-lite` es una versión rápida y económica, ideal para tareas simples; `gemini-3.5-flash` y `gemini-3.6-flash` son más capaces.
- Se accede a través de la **Google GenAI API** con una API Key.

### ¿Cómo obtener la API Key de Google?
1. Ir a [aistudio.google.com](https://aistudio.google.com)
2. Iniciar sesión con tu cuenta de Google
3. Click en **Get API Key** → **Create API Key**
4. Copiar la clave generada
5. En Colab, ir a **Secretos** (ícono de llave en el panel izquierdo)
6. Agregar un secreto con nombre `GOOGLE_API_KEY` y pegar la clave

## Paso 1: Configurar la API Key

```python
from google.colab import userdata
import os

# Obtener la API key desde los secretos de Colab
api_key = userdata.get('GOOGLE_API_KEY')

# Guardarla como variable de entorno
os.environ['GOOGLE_API_KEY'] = api_key

# Verificar que se haya cargado correctamente (nunca imprimas la clave, ni siquiera parcialmente)
print("API Key cargada:", "Sí" if api_key else "No")
```

## Paso 3: Crear el cliente y listar modelos disponibles

Creamos el cliente de Google GenAI con nuestra API Key y listamos los modelos Gemini disponibles.

```python
from google import genai

# Crear el cliente con la API Key
client = genai.Client(api_key=os.environ['GOOGLE_API_KEY'])

# Listar los modelos Gemini disponibles
print("Modelos Gemini disponibles:")
for m in client.models.list():
    if "gemini" in m.name:
        print(f"  - {m.name}")
```

## Paso 4: Workflow de exploración de oportunidades de IA

Vamos a crear un workflow donde Gemini nos ayuda a explorar oportunidades
de IA paso a paso. Cada paso usa la salida del anterior como contexto.

Para este ejemplo usaremos 3 modelos diferentes de Gemini (uno en cada paso), de menor a mayor capacidad:
`gemini-3.1-flash-lite`, `gemini-3.5-flash` y `gemini-3.6-flash`.

```python
# Paso 1 del workflow: Elegir un área de negocio
response_area = client.models.generate_content(
    model="gemini-3.1-flash-lite",
    contents="Elige un área de negocio que valga la pena explorar para una oportunidad de IA con Agentic."
)
area_negocio = (response_area.text or '').strip()

print("Paso 1 - Área de negocio sugerida por Gemini:")
print(area_negocio)
print("-" * 70)
```

```python
# Paso 2 del workflow: Identificar problemas en esa área
prompt_problemas = f"""Dado que el área de negocio elegida es:
{area_negocio}

Identifica 3 problemas principales que la IA podría resolver en esta área."""

response_problemas = client.models.generate_content(
    model="gemini-3.5-flash",
    contents=prompt_problemas
)
problemas = (response_problemas.text or '').strip()

print("Paso 2 - Problemas identificados:")
print(problemas)
print("-" * 70)
```

```python
# Paso 3 del workflow: Proponer soluciones con IA
prompt_soluciones = f"""Basándote en los siguientes problemas:
{problemas}

Propone una solución concreta usando IA para el problema más prometedor.
Incluye: tecnología sugerida, datos necesarios, y beneficio esperado."""

response_soluciones = client.models.generate_content(
    model="gemini-3.6-flash",
    contents=prompt_soluciones
)

print("Paso 3 - Solución propuesta:")
print((response_soluciones.text or '').strip())
```
