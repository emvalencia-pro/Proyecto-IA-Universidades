# ARBOLES DE DECISIÓN - Dataset Iris (clasificación flores)

---



Se debe crear un árbol de decisión para clasificar correctamente la variedad de la flor iris a partir de 150 muestras donde se tienen la información de:

* Ancho y largo de los pétalos
* sépalos.

Hay tres variedades de flor iris:

* 50 iris setosa
* 50 iris versicolor
* 50 iris virginica

## 📦 1. Cargar librerías

```python
import numpy as np
import pandas as pd

from sklearn.datasets import load_iris                      #dataset sobre la flor iris (uso libre)
from sklearn.model_selection import train_test_split        #clase para dividir dataset (train y test)
from sklearn.tree import DecisionTreeClassifier, plot_tree  #clase que permite implementar un arbol de desición y clase que permite dibujar el árbol
import matplotlib.pyplot as plt                             #libreria para generar graficos
```

## 📥 2. Cargar datos

```python
db_iris = load_iris()
```

## 🔣 3. Explorar datos

```python
#  Convertir los datos en un Dataframe para explorar datos
df_iris = pd.DataFrame (db_iris.data, columns= db_iris.feature_names)
df_iris['target'] = db_iris.target
df_iris.head()
```

## 🔍 4. Separar los datos ( train y test)

* Dividiremos el dataset en 80% train y 20% test, con el parámetro `test_size` .

* `random_state` controla la semilla del generador de números aleatorios que se utiliza para mezclar los datos antes de hacer la partición entre entrenamiento y prueba.

* Con `random_state=42`, los datos se mezclan siempre de la misma forma. La división entrenamiento/prueba será siempre la misma, y los resultados del modelo seran reproducibles. Este valor de `42` es una elección muy común por tradición en ciencia de datos.


```python
X, y = db_iris.data, db_iris.target
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.20, random_state=42)
```

## 🌳 5. Crear y Entrenar el Arbol de Decisión

```python
model = DecisionTreeClassifier (criterion='gini', max_depth=2, random_state=0)
model.fit(X_train, y_train)
```

## 📊 6. Visualizar el árbol

En cada nodo del gráfico, verás algo como esto:

* gini: índice de impureza en ese nodo.

* samples: número de ejemplos en el nodo.

* value: cuántos ejemplos hay de cada clase.

* class: clase mayoritaria (predicción).



```python
#para graficar Iris Setosa (0), - Iris Versicolour (1), - Iris Virginica (2)
plt.figure(figsize=(16, 12))
plot_tree(
    model,
    feature_names=db_iris.feature_names,
    class_names=['setosa', 'versicolor', 'virginica'],
    filled=True,
    rounded=True
)
plt.title("Árbol de Decisión para Iris (criterio: Gini)")
plt.show()
```

## ✅7. Evaluación del modelo

```python
from sklearn.metrics import accuracy_score
```

```python
# Realizar predicciones
y_pred= model.predict (X_test)
print(f"y_pred: {y_pred [0:10]}")
print(f"y_test: {y_test [0:10]}")
```

```python
# Calcular Exactitud
accuracy = accuracy_score(y_test, y_pred)
print(f"Accuracy del modelo: {accuracy:.2f}")
```

## ✅8. Análisis de Características (Features)

Para crear el árbol, el algoritmo encuentra que unas características son más importantes que otras...cuales son?

Para ello se usa la clase `feature_importances_`

*   Cuanto mayor es el valor de `feature_importances_`, más relevante fue esa columna para tomar decisiones dentro del árbol
*   Las importancias se calculan como la reducción promedio de la impureza (como Gini o entropía) producida por una característica en los nodos donde se usa.

```python
# Feature importance
feature_importance = model.feature_importances_
for i, importance in enumerate(feature_importance):
  print(f"Feature '{db_iris.feature_names[i]}': {importance:.4f}")
```

```python
# Visualize feature importance
plt.figure(figsize=(10, 6))
plt.bar(db_iris.feature_names, feature_importance)
plt.title("Feature Importance in Iris Dataset")
plt.xlabel("Features")
plt.ylabel("Importance")
plt.show()
```

## ✅9. Optimización del modelo

Es posible probar con diferentes niveles de profundidad y mirar que tan bueno es el modelo resultante de aplicar el algoritmo al observar como se comporta la metrica `Accuracy` del modelo.

```python
#la profundad de un árbol la medimos con max_depth
# creamos una lista con profundidades de 1 a 6
max_depth_lista = list (range(1, 6))
# creamos un listado de resultados de exactitud
accuracy= []
for depth in max_depth_lista:
  arbol_clasi= DecisionTreeClassifier(max_depth = depth, random_state = 0)
  arbol_clasi.fit(X_train, y_train)
  score = arbol_clasi.score (X_test, y_test)
  accuracy.append (score)
```

```python
# crear un grafico para ver los resultados
plt.plot(max_depth_lista, accuracy)
plt.ylabel('accuracy')
plt.xlabel('depth')
plt.grid(True)
plt.show()
```

```python
#para graficar el árbol optimizado: Iris Setosa (0), - Iris Versicolour (1), - Iris Virginica (2)
plt.figure(figsize=(16, 12))
plot_tree(
    arbol_clasi,
    feature_names=db_iris.feature_names,
    class_names=['setosa', 'versicolor', 'virginica'],
    filled=True,
    rounded=True
)
plt.title("Árbol de Decisión para Iris")
plt.show()
```
