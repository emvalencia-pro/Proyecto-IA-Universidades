# **Keras**: Clasificación binaria en una capa.

***

Ejemplo de una red neuronal de una sola capa.

**Puntos para tener en cuenta**


1. Se define un **modelo** usando el constructor `Sequential`. Esto permite agregar y configurar las capas del modelo.

2. Se pueden añadir capas para construir la arquitectura de la red neuronal artificial. La capa `Dense` es una capa cuyas neuronas están completamente conectadas entre si.

3. Es necesario especificar la dimensionalidad de la capa de entrada y de salida cuando se define la primera capa. En este ejemplo, el modelo tomará una dimensionalidad de 500 entradas y para producir una salida de dimensionalidad 1.

4. Se añade una función de activación, en este caso se usa la función Sigmoide.

5. Una vez se ha definido el modelo, éste se compila especificando parámetros tales como: función de coste, algoritmo de optimización y métricas que se desean calcular.

6. Se elige una función de coste de acuerdo a la tarea de clasificación. En este caso tenemos un problema de clasificación binaria, por lo tanto, se selecciona la función de coste: `binary_crossentropy`.

7. Se elige un algoritmo de optimización, en este caso se usa Stochastic Gradient Descent.

8. Compilado el modelo, se entrena y se evalúa.

# Importar librerías

```
import numpy as np
from keras.models import Sequential
from keras.layers import Dense, Activation
from keras.utils import plot_model as plot
```

#Definición del modelo

```
# Modelo de capas secuenciales
model = Sequential()
# Una sola capa densa: todas las neuronas interconectadas
model.add(Dense(1, input_dim=500, activation='sigmoid', name='capa1'))
# Compilar el modelo con los parámetros:
# optimización, funcion de coste y métrica a calcular
model.compile(optimizer='adam',
              loss='binary_crossentropy',
              metrics=['accuracy'])
# Resumen del modelo
model.summary()
```

# Datos que vamos a clasificar

Revisar los datos imprimiendo el contenido de las variables `X` y `y`. Comprender de qué se habla.

```
# Atributos: 1000 filas, 500 columnas
# variables predictoras
X = np.random.random((1000, 500))

# Etiquetas: 1000 filas, 1 columna
# clase, variable de salida
y = np.random.randint(2, size=(1000, 1)) # randint(2) Genera números enteros aleatorios entre 0 y 1.
```

#Evaluar model antes de entrenamiento

```
# evaluate(): Calcula la pérdida y las métricas configuradas.
score = model.evaluate(X, y, verbose=0) # El parámetro verbose=0 significa que no muestra mensajes en pantalla mientras evalúa
model_metrics = zip(model.metrics_names, score) # Emparejar el nombre de cada métrica con el valor obtenido. Ej: ('loss', 0.78) ('accuracy', 0.49)
model_metrics = set(model_metrics)
print('Antes del entrenamiento:', model_metrics)
```

**Nota:** algunas versiones actuales de Keras, `model.metrics_names` puede contener nombres agregados como **compile_metrics**, por lo que no siempre permite identificar individualmente todas las métricas.

# Entrenamiento del modelo

Cambie el parámetro `verbose=1` para ver la salida del proceso

```
history = model.fit(X, y, epochs=100, batch_size=32, verbose=0)
```

#Evaluar modelo después del entrenamiento


```
score = model.evaluate(X, y, verbose=0)
model_metrics = zip(model.metrics_names, score)
model_metrics = set(model_metrics)
print('Después del entrenamiento:', model_metrics)
```

# Gráficos para análisis


*   Optimización de la función de coste
*   Exactitud del modelo



```
import matplotlib.pyplot as plt
%matplotlib inline
hist = history.history
fig = plt.figure(figsize=(12, 5))
ax = fig.add_subplot(1, 2, 1)
ax.plot(hist['loss'], lw=3)
ax.set_title('Training loss', size=15)
ax.set_xlabel('Epoch', size=15)
ax.tick_params(axis='both', which='major', labelsize=15)
ax = fig.add_subplot(1, 2, 2)
ax.plot(hist['accuracy'], lw=3)
ax.set_title('Training accuracy', size=15)
ax.set_xlabel('Epoch', size=15)
ax.tick_params(axis='both', which='major', labelsize=15)
plt.show()
```
