## **PERCEPTRON**

```python
# =======================================
# 📚 Importación de librerías necesarias
# =======================================
import numpy as np
from sklearn.datasets import make_classification
```

`make_classification` crea un dataset sintético de características numéricas junto con etiquetas de clase (0, 1, etc.), simulando un problema de clasificación binaria o multiclase.

```python
# =============================================
# Fijamos la semilla 🌱
# =============================================

np.random.seed(42)
```

```python
# =======================================
# 📊 Datos de entrada (X) y etiquetas (y)
# =======================================


X, y = make_classification(
    n_samples=500,         # 500 ejemplos
    n_features=2,          # 2 características X₁ y X₂.(perfecto para visualizar)
    n_redundant=0,         # características redundantes (combinaciones lineales de las informativas)
    n_informative=2,       # número de características útiles para la clasificación
    n_clusters_per_class=1,# Número de agrupamientos necesarios para generar cada clase.
    n_classes=2,           # Número de clases (2 por defecto: 0 y 1).
    random_state=42        # Fija la semilla aleatoria para obtener resultados reproducibles.
)

# Aseguramos que y sea columna
y = y.reshape(-1, 1)
```

```python
# Inicialización
n_features = X.shape[1]
w = np.zeros((n_features, 1))  # (n, 1) Pesos (w) de la red
b = 0                          # sesgo b
lr = 0.01                      # tasa de aprendizaje
epochs = 750

# Función sigmoide
def sigmoid(z):
    return 1 / (1 + np.exp(-z))
```

```python
# 🔄 Entrenamiento del perceptrón
for epoch in range(epochs):
    z = np.dot(X, w) + b         # (m, 1) Cálculo de funcion de trasferencia z
    y_hat = sigmoid(z)           # (m, 1) Predicción de y
    error = y_hat - y            # (m, 1) Cálculo del error

    # Calcular los gradientes respecto al peso (w) y el sesgo (b)
    dw = np.dot(X.T, error) / X.shape[0]  # (n, 1)
    db = np.mean(error)                   # escalar

    # Actualización de los pesos (w) y el sesgo (b)
    w = w - lr * dw
    b = b - lr * db

    # Cálculo de la pérdida (cada 100 epocas: impresión epoca y perdida)
    if epoch % 100 == 0:
        loss = -np.mean(y * np.log(y_hat + 1e-8) + (1 - y) * np.log(1 - y_hat + 1e-8))
        print(f"Época {epoch} - Pérdida: {loss:.4f}")
```

```python
# =====================
# 🔍 Predicción final
# =====================
z_final = np.dot(X, w) + b
y_pred = (sigmoid(z_final) > 0.5).astype(int)

#y_pred = (sigmoid(z_final) > 0.5).astype(int) Si prefieren 0 y 1 como enteros

print("\nPredicciones finales:")
print(y_pred.ravel())

print("\nEtiquetas reales:")
print(y.ravel())
```
